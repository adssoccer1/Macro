# Exploring Concurrent Necessity and Today's Macro Landscape

For a trader, understanding the principle of concurrent necessity, as described by Alex Gurevich in *The Next Perfect Trade*, is essential to avoiding trades "traped in the bowels of investment mediocrity". 

Concurrent necessity is the idea that by the time event A is occurring, event B must already be underway. For example, in many economic environments, when U.S. interest rate differentials versus other currencies decline, the dollar weakens concurrently. Taking this a step further, trade A is inferior to trade B if trade A implies B must of already happened. Another example being, if you wanted to bet on an economic slowdown, then going long bonds almost always dominates shorting stocks. This is because falling rates are a concurrent necessity with respect to equity weakness. In other words, there are more scenarios in which yields decline than in which equities fall, making long bonds the dominant trade.

Avoid inferior trades. Superior trades succeed across a broader range of possible outcomes.

Building on Gurevich’s framework of concurrent necessity, two additional principles become essential: the *economic causality chain* and *perfect trades*.

The economic causality chain refers to parsing the macro environment into cause-and-effect linkages that drive movements between asset classes. Gurevich illustrates this with a simple interest rate–equity chain, a mental model most traders implicitly carry. By examining such chains, one can identify where concurrent necessities exist between assets and use that insight to select dominant trade 



This basic causality chain is too simplistic for the post-COVID era and would have misled traders by suggesting that elevated nominal and real rates should have suppressed equities after the Fed's final 2023 rate hike. Inflation peaked in June 2022, at which point the combination of disinflation and rate hikes steadily increased real rates creating a feedback loop in which disinflation, increased real rates further, which puts more downward pressure on inflation. 



However, this loop was arrested before inflation returned to target leaving real yields historically high around 2% as U.S. stocks continue hitting new all-time highs. Below is my attempt to map the post-COVID macro landscape to an updated causality chain:



Notice how a series of unconventional stimuli replaced the traditional role of rate cuts (box 10), allowing equities to rebound from their 2022 lows and sustain momentum even in the face of restrictive real yields:

- Treasury supply management and liquidity drains
    - Secretary Yellen’s shift toward heavier issuance of short-dated bills, combined with the drawdown of the Fed’s reverse repo facility, recycled liquidity back into risk assets rather than crowding out private credit. This effectively eased financial conditions without a policy rate cut.
- Fiscal dominance in the post-COVID regime
    - Despite higher rates, household net interest burdens remain contained, as much consumer and corporate debt was locked in at lower coupons. At the same time, persistent federal deficits—running near 6–7% of GDP—have pushed excess capital out the risk curve, forcing investors into equities and credit even as Treasuries offered higher nominal yields. Pandemic-era savings, though diminished, still provided residual support through 2023–24.
- Robust consumer demand
    - Wage growth has outpaced inflation in key sectors, keeping real incomes positive on a trend basis. Coupled with strong household balance sheets and wealth effects from rising asset prices, consumption stayed resilient despite tighter policy.
- AI-driven investment cycle
    - Corporate capex has been buoyed by the artificial intelligence boom, creating a new source of demand that offsets traditional cyclical slowdowns. This has reinforced equity valuations even as financing costs rose.
- Policy flexibility at the margin
    - Rate cuts still occur episodically during growth scares, preventing a full tightening cycle from breaking the expansion. These tactical easings have provided downside insurance to risk assets.

These stimuli have so far blunted the impact of high real rates which remains at elevated levels:

Today is a good time to draw one's own thesis for where the U.S. economy is headed. Strong arguments can be made for multiple paths forward:
- An inflationary “crash up”, where over easy fiscal and monetary policy overwhelm restrictive rates, driving asset prices higher. 
- A deflationary “crash down”, where real rates finally dominate, leading to weaker consumption, falling employment, and lower asset prices.
- Or some muddled middle path, where growth slows but stimulus and investment cycles prevent outright contraction and the gap between asset prices and the real econonmy gradually narrows. 

To bet on a slowdown, the dominant trade is long bonds, and perhaps the most dominant is long TIPS. The inflation protection ensures that if the slowdown fails to materialize and inflation instead reasserts itself, the trade can still succeed.

To bet on a economic strength the dominant trade is not to short bonds, it's to bet on long stocks. It's perfectly plausible rate cuts are what drive the next leg of U.S. economic growth and stock performance. 

However, a new force must be added to the causality chain: a weakening U.S. dollar.

My paper on Modern Mercantilism [here](https://github.com/adssoccer1/Macro/blob/main/Modern%20Mercantilism%20Challenge/Aidan_Sinclair_A_Reflexive_Rebalancing.pdf) outline the case for sustained dollar weakness, driven by several forces:
 - Trump administration wants a weaker dollar to reduce the U.S. Capital Account surplus.
 - Export-surplus economies with overextended Net International Investment Positions (NIIPs) are positioned for reflexive inflows back into domestic assets. As these flows reverse, their currencies strengthen relative to the dollar.
 - In an U.S. inflationary crash-up, overly loose fiscal and monetary policy push the U.S. economy into an inflationary boom, the dollar is likely to depreciate relative to real assets and the currencies of more disciplined economies.
 - In a deflationary U.S. slowdown, the U.S. will lean on more fiscal stimulus. Unlike in past cycles, foreign capital may no longer automatically flow into U.S. assets during global risk-off events, adding another scenario for dollar weakness.

In short, export-surplus countries are could experience a reflexive currency appreciation as the Trump administration actively pursues a weaker dollar to rebalance U.S. trade. 

Consider South Korea, where exports to the U.S. account for 7% of GDP. An appreciating won could reduce American demand for Korean goods, encouraging Korea to reorient its export-driven economy. In such a scenario, fiscal stimulus is appropriate, especially considering Korea's relative post-COVID fiscal responsibility. Korean officials have already signaled willingness to support growth through government spending.

Futhermore, South Korea's over extended NIIP and overly large FX reserves offer potential energy for the start of what could be a reflexive repatration of capital into Korean assets. Add that this in itself will strengthen the Won taming inflation allowing for continued stimulus.

For traders, Monitoring Korea’s trade balance and the announcement of new fiscal stimulus cycle offers the chance to capture this reflexive shift. Long Korean equities is an exciting trade if this reflexive match is ignited. 

Let us now return to the second principle that builds on the idea of concurrent necessity: the perfect trade. Gurevich defines a perfect trade as a pairing of two dominant trades, A and B, in which neither has another trade that dominates it through concurrent necessity. Additionally, trade A must be a concurrent necessity with respect to the failure of trade B, and trade B must be a concurrent necessity with respect to the failure of trade A.

A straightforward expression of my weakening U.S. dollar view is to go short USD:KRW. This position is a dominant trade: it directly captures the structural forces driving dollar weakness, and no other trade is strictly necessary for its success.

A second expression would be to go long Korean equities. This trade would benefit from fiscal stimulus in Korea, which is likely if an appreciating Won pressures exports. In addition, fiscal expansion combined with capital repatriation could ignite a reflexive cycle of equity outperformance. However, this position is dominated by the short USDKRW trade. If Korea chooses not to stimulate and instead accepts export retrenchment, the long quity thesis falls apart. I like long Korean equities as a speculative trade nonetheless. 

Given the short-dollar position, can we identify a complementary trade that could complete a *perfect trade*? Well, what is a concurrent necessity to its failure (a strong U.S. dollar)? We know periods of dollar strength have often coincided with higher rates and I see this persisting in 2025. But what must have already happened for short rates to rise? Equities must already be going up! This makes long U.S. equities a concurrent necessity with respect an increasing U.S. dollar. 

Now, what is the concurrent necessity of weaker equities? Despite impressive equity performance in 2025 the STIR market is fast to price in cuts during equity declines. Lower short rates are a concurrent necessity with respect to a weaker U.S. equity market. And, in turn, as rates fall, the U.S. dollar weakens making the short-dollar position dominant to short equites. 

Through this lens, shorting the dollar and going long U.S. equities together approximate the structure of a perfect trade:
- If equities fall -> rates are cut -> the dollar weakens, supporting the short-USD leg.
- If the dollar strengthens -> it is typically on the back of higher rates -> which necessarily coincides with rising equities, supporting the long-equity leg.
