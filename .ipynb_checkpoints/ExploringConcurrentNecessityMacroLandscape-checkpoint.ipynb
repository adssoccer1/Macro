# Exploring Concurrent Necessity and Today's Macro Landscape

For a trader, understanding the principle of concurrent necessity, as described by Alex Gurevich in *The Next Perfect Trade*, is essential to avoiding trades "trapped in the bowels of investment mediocrity". 

Concurrent necessity is the idea that by the time event A is occurring, event B must already be underway. For example, in many economic environments, when U.S. interest rate differentials versus other currencies decline, the dollar weakens concurrently. Taking this a step further, trade A is inferior to trade B if trade A implies B must have already happened. Another example being, if you wanted to bet on an economic slowdown, then going long bonds almost always dominates shorting stocks. This is because falling rates are almost always a concurrent necessity with respect to equity weakness. In other words, there are more scenarios in which yields decline than in which equities fall, making long bonds the dominant trade (See Appendix A1).

These inter-asset-class relationships rest on assumptions that can change across different environments. What must remain constant for long-term outperformance is the discipline to avoid inferior trades. Superior trades succeed across a broader range of possible outcomes.

Building on Gurevich’s framework of concurrent necessity, two additional principles become essential: the *economic causality chain* and *perfect trades*.

### The Causality Chain

The economic causality chain refers to parsing the macro environment into cause-and-effect linkages that drive movements between asset classes. Gurevich illustrates this with a simple interest rate–equity chain, a mental model most traders implicitly carry. By examining such chains, one can identify the concurrent necessities that exist between assets (See Appendix A1).



This basic causality chain is too simplistic for the post-COVID era and would have misled traders by suggesting that elevated nominal and real rates should have suppressed equities after the Fed's final 2023 rate hike. Inflation peaked in June 2022, at which point the combination of disinflation and rate hikes steadily increased real rates creating a feedback loop in which disinflation, increased real rates further, which puts more downward pressure on inflation. Disinflation was self-reinforcing.



However, this loop was arrested before inflation returned to target, leaving real yields historically high around 2% as U.S. stocks continue hitting new all-time highs. Below is my attempt to map the post-COVID macro landscape to an updated causality chain:



### Today's Macro Landscape

Notice how a series of unconventional stimuli replaced the traditional role of rate cuts (box 10), allowing equities to rebound from their 2022 lows and sustain momentum even in the face of restrictive real yields:

- Financial engineering [which I wrote about here](https://github.com/adssoccer1/Macro/blob/main/2023DeathOfVol.ipynb)
    - Secretary Yellen’s shift toward heavier issuance of short-dated bills, combined with the drawdown of the Fed’s reverse repo facility, pushed liquidity back into risk assets rather than crowding out private credit. This effectively eased financial conditions without a policy rate cut.
- Fiscal dominance post-COVID
    - Despite higher rates, household net interest burdens remain contained, as much consumer and corporate debt was locked in at lower coupons. At the same time, persistent federal deficits—running near 6–7% of GDP—have pushed excess capital out the risk curve, forcing investors into equities and credit even as Treasuries offered higher nominal yields. Pandemic stimulus and savings, though diminished, still provided residual support.
- Robust consumer
    - Wage growth has outpaced inflation, keeping real incomes elevated. Coupled with strong household balance sheets and wealth effects from rising asset prices, consumption stayed strong.
- AI Capex boom
    - Corporate capex has been buoyed by the artificial intelligence boom, creating a new source of demand that offsets traditional cyclical slowdowns. This has reinforced equity valuations even as financing costs rose.
- Rate cuts
    - Mini rate cutting cycles still occur during growth scares like last September, preventing a full tightening cycle from breaking the expansion.

These stimuli have so far blunted the impact of high real rates which remain at elevated levels (See Appendix A2).

Today is a good time to draw one's own thesis for where the U.S. economy is headed. Strong arguments can be made for different paths forward:
- An inflationary “crash up”, where over easy fiscal and monetary policy overwhelm restrictive rates, driving asset prices higher. 
- A deflationary “crash down”, where real rates finally bite, leading to weaker consumption, falling employment, and lower asset prices.
- Or some muddled middle path, where growth slows but stimulus and investment cycles prevent outright contraction and the gap between asset prices and the real economy gradually narrows. 

To bet on a slowdown, the dominant trade is long bonds.

To bet on economic strength the dominant trade is not to short bonds, it's to bet on long stocks. It's perfectly plausible rate cuts are what drive the next leg of U.S. economic growth and stock performance. 

However, a new force must be added to the causality chain: a weakening U.S. dollar.

My paper on Modern Mercantilism ([Read Here](https://github.com/adssoccer1/Macro/blob/main/Modern%20Mercantilism%20Challenge/Aidan_Sinclair_A_Reflexive_Rebalancing.pdf)) outlines the case for sustained dollar weakness, driven by several forces:
 - Trump administration wants a weaker dollar to reduce the U.S. Current Account deficit and Capital Account surplus.
 - Export-surplus economies with overextended Net International Investment Positions (NIIPs) are positioned for reflexive inflows back into domestic assets. As these flows reverse, their currencies strengthen relative to the dollar.
 - In a U.S. inflationary crash-up, overly loose fiscal and monetary policy push the U.S. economy into an inflationary boom, the dollar is likely to depreciate relative to real assets and currencies of more disciplined economies.
 - In a deflationary U.S. slowdown, the U.S. will lean on more fiscal stimulus which will be dollar negative. Plus, unlike in past cycles, foreign capital may no longer automatically flow into U.S. assets during global risk-off events, adding another scenario for dollar weakness.

In short, export-surplus countries could experience a reflexive currency appreciation as the Trump administration actively pursues a weaker dollar to rebalance U.S. trade. 

Consider South Korea, where exports to the U.S. account for 7% of GDP. An appreciating Won could reduce American demand for Korean goods, encouraging Korea to reorient its export-driven economy. In such a scenario, fiscal stimulus is appropriate, especially considering Korea's relative post-COVID fiscal responsibility. Korean officials have already signaled willingness to support growth through government spending (See Appendix A3).

Furthermore, South Korea's overextended NIIP and overly large FX reserves offer potential energy for the start of what could be a reflexive repatriation of capital into Korean assets ([read more here](https://github.com/adssoccer1/Macro/blob/main/Modern%20Mercantilism%20Challenge/Aidan_Sinclair_A_Reflexive_Rebalancing.pdf)). This in itself will strengthen the Won, limiting inflation, allowing for continued stimulus.

For traders, monitoring Korea’s trade balance and the announcement of new fiscal stimulus cycle offers the chance to capture this reflexive shift. Long Korean equities is an exciting trade if this reflexive match is ignited (See Appendix A3). 

### A Perfect Trade?

Let us now return to the second principle that builds on the idea of concurrent necessity: the perfect trade. Gurevich defines a perfect trade as a pairing of two superior trades, A and B, in which neither has another trade that dominates it through concurrent necessity. Additionally, trade A must be a concurrent necessity with respect to the failure of trade B, and trade B must be a concurrent necessity with respect to the failure of trade A.

A straightforward expression of my weakening U.S. dollar view is to go short USD:KRW. This position is a dominant trade: it directly captures the structural forces driving dollar weakness, and no other trade is strictly necessary for its success.

A second expression would be to go long Korean equities. This trade would benefit from fiscal stimulus in Korea, which is likely if an appreciating Won pressures exports. In addition, fiscal expansion combined with capital repatriation could ignite a reflexive cycle of equity outperformance. However, this position is dominated by the short USD:KRW trade. If Korea chooses not to stimulate and instead retrenches in their export oriented economy, the long equity thesis falls apart. I like long Korean equities as a speculative trade nonetheless. 

Assuming trade A is the short-dollar, long Won position, can we identify a complementary trade that completes a *perfect trade*? Well, what is a concurrent necessity to its failure (a strong U.S. dollar)? We know periods of dollar strength have often coincided with higher rates and I see this persisting in 2025. But what must have already happened for short rates to rise? Equities must already be going up! This makes long U.S. equities a concurrent necessity with respect to an increasing U.S. dollar. 

Now, what is the concurrent necessity of weaker equities (the failure of trade B)? Despite impressive equity performance in 2025 the STIR market is fast to price in cuts during equity declines. Lower short rates are a concurrent necessity with respect to a weaker U.S. equity market. And, in turn, as rates fall, the U.S. dollar weakens making the short-dollar position dominant to short equities. 

Through this lens, shorting the dollar and going long U.S. equities together, approximate the structure of a perfect trade and my current thesis:
- If equities fall -> rates will be are cut -> the dollar weakens, supporting the short-USD leg.
- If the dollar strengthens -> it will be on the back of higher rates -> which necessarily coincides with rising equities, supporting the long-equity leg (See Appendix A1).

### Some Nuance

A critical point is that causality chains shift across different economic environments. For example, 2022 breaks, the assumption I hold today, that rising equities are a concurrent necessity with respect to higher rates. That year, rates rose  while equities fell. Concurrent necessity is not timeless; it must be recalibrated to the prevailing macro regime.

Furthermore, Gurevich emphasizes that a perfect trade must be built from two superior trades. He defines a superior trade as one that:
- It is aligned with a secular trend.
- It carries positively.
- It is supported by terminal value expectation.
- It is aligned with historical pattern.
- It is aligned with global growth.
- It is executed with a clear long-term strategy.
- It has a small downside.
- It has high chances of succeeding even when underlying view is wrong.
- It is not dominated by other trades.

Through this lens, the two legs of my proposed “perfect trade” deserve scrutiny. Going long U.S. equities plausibly fits the superior trade criteria. However, shorting the dollar does not carry positively, nor is it aligned with a secular trend. Shorting the dollar is in part a bet on a major change in global trade and capital flows, based on my Modern Mercantilism thesis (See Appendix A3). 

This raises an important nuance: trades based on anticipated structural breaks, such as a shift in the dollar’s role, do not meet Gurevich’s strict definition of superiority, even if they are dominant relative to alternatives. The implication is that my proposed “perfect trade” structure is closer to a conditional trade: it satisfies the concurrent necessity conditions, but its validity ultimately rests on the correctness of the structural view that U.S. dollar weakness is not a cyclical blip but the beginning of a new secular phase.

### A1 Concurrent Necessity and Conditional Probability

Concurrent necessity between asset classes may sound similar to correlation, but it is more precisely captured through conditional probability. Correlation measures whether two variables tend to move together in general, but it does not tell us the likelihood that one movement occurs given the other has already happened. Conditional probability, by contrast, allows us to quantify directionality: if equities decline, what is the probability that short-term yields also decline? This approach focuses on sufficiency rather than symmetry, highlighting dependencies that underlie economic causality chains. ### A3 A Reflexive Rebalancing into Korea?

*Below is an excerpt from my piece on Modern Mercantilism, [linked here](https://github.com/adssoccer1/Macro/blob/main/Modern%20Mercantilism%20Challenge/Aidan_Sinclair_A_Reflexive_Rebalancing.pdf)*

Some surplus countries are in a unique position to embrace shifts in trade and currencies in ways that can transform their economies and raise living standards. For instance, relative fiscal responsibility post-Covid has given South Korea a low debt-GDP ratio, 10 year yield, a +30bps 10-2s spread, and a whopping 24% FX reserves/GDP. Most noticeably, Korea’s NIIP/GDP has increased by 965% over just the last 10 years. 


As Korean exports to the U.S., which account for greater than 7% of GDP, come under pressure due to a weakening USD (and strengthening Won), monetary and fiscal support to promote domestic consumption and a smooth transition is appropriate. This is especially warranted because Korea is in a position to offset upward pressure on inflation and interest rates due to several unique reasons:
- A large coffer of FX reserves can be focused on domestic initiatives without the need for issuing new debt to fund expenditures. 
- High domestic savings rate (~35%) can easily absorb an increase in KTB issuance. 
- Appreciation of the Won will decrease import costs of raw materials and goods countering inflationary pressures. 
- The explosion of Korea’s NIIP over the last 10 years has the potential to repatriate, keeping funding costs low and pushing out capital on the risk curve. 

The combination of these forces could then attract foreign inflows, increasing demand for the Won, leading to an exciting virtuous cycle similar to that experienced in the U.S.

More granularly, it’s likely that within the next 18 months, South Korea will begin to stimulate preemptively to counter the effects of a weakening U.S. dollar. This will manifest through a steepening of the Korean Treasury Bond curve. A steeper curve will initiate repatriation of foreign portfolio holdings, particularly given Korea’s large NIIP. 

Crucially, surplus nations no longer have incentives to accumulate outsized FX reserves or paper assets from large DMs. What has previously served to stabilize currency pegs and protect against sovereign debt crises no longer makes sense. Why, after years of sovereign bond drawdowns, inflationary policies and promises of future debasement, would surplus countries continue to buy the increasingly worthless assets of DMs instead of investing in real assets? There is an accelerating global shift toward converting the abundant paper reserves and lopsided NIIPs into tangible investments: commodities, domestic infrastructure, military capacity, and local supply chains. 

Korea is particularly advantaged by its substantial reserve of paper claims that it can sell for domestic investment. As capital flows home, the KRW will strengthen further, reinforcing disinflationary forces and lowering the real cost of imports. These flows will likely continue to push Korean rates higher, not because of instability, but because of increasing expectations for growth.

Higher real rates in Korea, even potentially relative to the U.S., will begin to attract foreign inflows from global investors. Korea’s fiscal injections, currency strength, and healthy balance sheet give it an unusual ability to sustain expansionary policy without the usual penalty of currency depreciation or rising credit risk. This is a virtuous cycle and offers a reflexive opportunity if Korean policymakers embrace it: repatriation boosts domestic markets, rising asset prices attract foreign flows, the Won appreciates further, disinflation provides more fiscal headroom, and the cycle repeats.

```python
# pip install yfinance pandas_datareader matplotlib pandas numpy

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import pandas_datareader.data as pdr
import yfinance as yf

def _force_series(x, name):
    if isinstance(x, pd.DataFrame):
        x = x.iloc[:, 0]
    x = pd.Series(x)
    x.name = name
    return x.astype(float)

def load_monthly(start='1985-01-01'):
    y10 = pdr.DataReader('DGS10', 'fred', start=start)
    y10 = _force_series(y10, 'DGS10_pct').resample('ME').last()

    spx_raw = yf.download('^GSPC', start=start, progress=False, auto_adjust=True)
    spx = _force_series(spx_raw['Close'], 'SPX').resample('ME').last()

    return spx, y10

spx, y10 = load_monthly()

spx_fwd_24m = (spx.shift(-24) / spx - 1.0).rename('SPX_fwd_24m_pct')    
y10_back_24m = (y10.shift(24) - y10).rename('DGS10_back_24m_pp')           

last_orange = spx_fwd_24m.dropna().index.max()
last_blue   = y10_back_24m.dropna().index.max()

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(spx_fwd_24m.dropna().index,
         spx_fwd_24m.dropna().values * 100.0,
         color='orange', linewidth=1.8,
         label='2Y Forward-Looking SPX % Change')

ax1.set_ylabel('2Y Forward-Looking SPX Differential (%)')
ax1.set_xlabel('Date')
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(y10_back_24m.dropna().index,
         y10_back_24m.dropna().values,
         color='blue', linewidth=1.8,
         label='2Y Backward-Looking 10Y Yield Differential (pp)')
ax2.set_ylabel('2Y Backward-Looking 10Y Yield Differential (pp)')

ax1.set_title('2Y Backward 10Y Yield Differential vs 2Y Forward SPX % Change (Monthly)')

if (pd.notna(last_orange)) and (pd.notna(last_blue)) and (last_blue > last_orange):
    ax1.axvspan(last_orange, last_blue, alpha=0.08)

l1, lab1 = ax1.get_legend_handles_labels()
l2, lab2 = ax2.get_legend_handles_labels()
ax1.legend(l1 + l2, lab1 + lab2, loc='upper left')

plt.show()

print(f"Blue series (yields) through: {last_blue.date()}")
print(f"Orange series (forward SPX) through: {last_orange.date()}  (stops ~24m earlier by construction)")
```

```python

import pandas as pd
import numpy as np
import yfinance as yf
from pandas_datareader import data as pdr
import datetime

START = "2024-01-01"
END   = datetime.date.today().strftime("%Y-%m-%d")
EQ_TICKER = "SPY"

eq_raw = yf.download(EQ_TICKER, start=START, end=END, progress=False, auto_adjust=True)
if isinstance(eq_raw.columns, pd.MultiIndex):
    eq_raw.columns = eq_raw.columns.get_level_values(0)
if "Close" in eq_raw.columns:
    equity = eq_raw[["Close"]].rename(columns={"Close":"equity"}).dropna()
elif "Adj Close" in eq_raw.columns:
    equity = eq_raw[["Adj Close"]].rename(columns={"Adj Close":"equity"}).dropna()
else:
    raise KeyError(f"No 'Close' or 'Adj Close' in columns: {list(eq_raw.columns)}")

y10 = pdr.DataReader("DGS10",  "fred", START, END).rename(columns={"DGS10":"y10"}).dropna()
y6m = pdr.DataReader("DGS6MO", "fred", START, END).rename(columns={"DGS6MO":"y6m"}).dropna()
usd = pdr.DataReader("DTWEXBGS", "fred", START, END).rename(columns={"DTWEXBGS":"usd"}).dropna()

eq_m  = equity.resample("M").last()
y10_m = y10.resample("M").last()
y6m_m = y6m.resample("M").last()
usd_m = usd.resample("M").last()

df = pd.concat([eq_m, y10_m, y6m_m, usd_m], axis=1, join="inner").dropna()
df.columns = ["equity","y10","y6m","usd"]

df["ret_equity_pct"] = df["equity"].pct_change() * 100.0
df["chg_y10_bp"]     = df["y10"].diff() * 100.0
df["chg_y6m_bp"]     = df["y6m"].diff() * 100.0
df["ret_usd_pct"]    = df["usd"].pct_change() * 100.0
df = df.dropna().reset_index().rename(columns={"index":"date"})

df["A_eq_dn_1pct"] = (df["ret_equity_pct"] <= -1.0).astype(int)   # ↓ ≥1% m/m
df["A_eq_up_1pct"] = (df["ret_equity_pct"] >= +1.0).astype(int)   # ↑ ≥1% m/m

df["A_y10_up"]  = (df["chg_y10_bp"] > 0.0).astype(int)
df["A_y10_dn"]  = (df["chg_y10_bp"] < 0.0).astype(int)
df["A_y6m_up"]  = (df["chg_y6m_bp"] > 0.0).astype(int)
df["A_y6m_dn"]  = (df["chg_y6m_bp"] < 0.0).astype(int)

df["B_usd_up"]  = (df["ret_usd_pct"] > 0.0).astype(int)
df["B_usd_dn"]  = (df["ret_usd_pct"] < 0.0).astype(int)

for col in ["B_usd_up","B_usd_dn","A_y10_up","A_y6m_up","A_y10_dn","A_y6m_dn", "A_eq_up_1pct","A_eq_dn_1pct"]:
    df[col + "_t1"] = df[col].shift(-1)

# ----- CN helper -----
def cn_stats(frame, A_col, B_col):
    sub = frame.dropna(subset=[A_col, B_col])
    A = sub[A_col] == 1
    B = sub[B_col] == 1
    n  = len(sub)
    nA = int(A.sum()); nB = int(B.sum())
    nAB = int((A & B).sum())
    pA = nA/n if n>0 else np.nan
    pB = nB/n if n>0 else np.nan
    pB_given_A = (nAB/nA) if nA>0 else np.nan
    pA_given_B = (nAB/nB) if nB>0 else np.nan
    eps = 1e-12
    lift = pB_given_A/(pB+eps) if pd.notna(pB_given_A) and pd.notna(pB) else np.nan
    nA_notB      = nA - nAB
    nNotA_B      = nB - nAB
    nNotA_notB   = n - (nAB + nA_notB + nNotA_B)
    odds_ratio   = ((nAB+eps)*(nNotA_notB+eps))/((nA_notB+eps)*(nNotA_B+eps))
    return pd.Series({
        "N": n, "N(A)": nA, "N(B)": nB, "N(A∩B)": nAB,
        "P(A)": pA, "P(B)": pB,
        "P(B|A)": pB_given_A, "P(A|B)": pA_given_B,
        "Lift": lift, "Odds Ratio": odds_ratio
    })

rows = []

def add_row(Acol, Bcol, label, horizon):
    s = cn_stats(df, Acol, Bcol).to_dict()
    s["Pair"] = label
    s["Horizon"] = horizon
    rows.append(s)

add_row("A_eq_up_1pct", "A_y10_up",        "Equities↑(≥+1% m/m) ⇒ 10y↑", "t")
add_row("A_eq_up_1pct", "A_y10_up_t1",     "Equities↑(≥+1% m/m) ⇒ 10y↑", "t+1")
add_row("A_eq_up_1pct", "A_y6m_up",        "Equities↑(≥+1% m/m) ⇒ 6m↑",  "t")
add_row("A_eq_up_1pct", "A_y6m_up_t1",     "Equities↑(≥+1% m/m) ⇒ 6m↑",  "t+1")

add_row("A_y10_dn",     "B_usd_dn",        "10y↓ ⇒ USD↓",                "t")
add_row("A_y10_dn",     "B_usd_dn_t1",     "10y↓ ⇒ USD↓",                "t+1")
add_row("A_y6m_dn",     "B_usd_dn",        "6m↓ ⇒ USD↓",                 "t")
add_row("A_y6m_dn",     "B_usd_dn_t1",     "6m↓ ⇒ USD↓",                 "t+1")

add_row("A_y10_up", "A_eq_up_1pct",      "10y↑ ⇒ Equities↑(≥+1% m/m)", "t")
add_row("A_y10_up", "A_eq_up_1pct_t1",   "10y↑ ⇒ Equities↑(≥+1% m/m)", "t+1")

add_row("A_y6m_up", "A_eq_up_1pct",      "6m↑ ⇒ Equities↑(≥+1% m/m)",  "t")
add_row("A_y6m_up", "A_eq_up_1pct_t1",   "6m↑ ⇒ Equities↑(≥+1% m/m)",  "t+1")

add_row("A_eq_dn_1pct", "B_usd_dn",    "Equities↓(≤−1% m/m) ⇒ USD↓", "t")
add_row("A_eq_dn_1pct", "B_usd_dn_t1", "Equities↓(≤−1% m/m) ⇒ USD↓", "t+1")

add_row("A_eq_dn_1pct", "A_y6m_dn",    "Equities↓(≤−1% m/m) ⇒ 6m↓",  "t")
add_row("A_eq_dn_1pct", "A_y6m_dn_t1", "Equities↓(≤−1% m/m) ⇒ 6m↓",  "t+1")

add_row("A_eq_dn_1pct", "B_usd_up",        "Equities↓(≤−1% m/m) ⇒ USD↑", "t")
add_row("A_eq_dn_1pct", "B_usd_up_t1",     "Equities↓(≤−1% m/m) ⇒ USD↑", "t+1")
add_row("A_y10_up",     "B_usd_up",        "10y↑ ⇒ USD↑",                "t")
add_row("A_y10_up",     "B_usd_up_t1",     "10y↑ ⇒ USD↑",                "t+1")
add_row("A_y6m_up",     "B_usd_up",        "6m↑ ⇒ USD↑",                 "t")
add_row("A_y6m_up",     "B_usd_up_t1",     "6m↑ ⇒ USD↑",                 "t+1")

add_row("B_usd_dn", "A_y10_dn", "USD↓ ⇒ 10y↓", "t")
add_row("B_usd_dn", "A_y10_dn_t1", "USD↓ ⇒ 10y↓", "t+1")
add_row("B_usd_dn", "A_y6m_dn", "USD↓ ⇒ 6m↓", "t")
add_row("B_usd_dn", "A_y6m_dn_t1", "USD↓ ⇒ 6m↓", "t+1")

add_row("B_usd_dn", "A_y10_up", "USD↓ ⇒ 10y↑", "t")
add_row("B_usd_dn", "A_y10_up_t1", "USD↓ ⇒ 10y↑", "t+1")
add_row("B_usd_dn", "A_y6m_up", "USD↓ ⇒ 6m↑", "t")
add_row("B_usd_dn", "A_y6m_up_t1", "USD↓ ⇒ 6m↑", "t+1")

add_row("B_usd_up", "A_y10_up",    "USD↑ ⇒ 10y↑", "t")
add_row("B_usd_up", "A_y10_up_t1", "USD↑ ⇒ 10y↑", "t+1")
add_row("B_usd_up", "A_y10_dn",    "USD↑ ⇒ 10y↓", "t")
add_row("B_usd_up", "A_y10_dn_t1", "USD↑ ⇒ 10y↓", "t+1")

add_row("B_usd_up", "A_y6m_up",    "USD↑ ⇒ 6m↑", "t")
add_row("B_usd_up", "A_y6m_up_t1", "USD↑ ⇒ 6m↑", "t+1")
add_row("B_usd_up", "A_y6m_dn",    "USD↑ ⇒ 6m↓", "t")
add_row("B_usd_up", "A_y6m_dn_t1", "USD↑ ⇒ 6m↓", "t+1")

out = pd.DataFrame(rows)[
    ["Pair","Horizon","N","N(A)","N(B)","N(A∩B)","P(A)","P(B)","P(B|A)","P(A|B)","Lift","Odds Ratio"]
].sort_values(["Pair","Horizon"]).reset_index(drop=True)

with pd.option_context('display.float_format', '{:.3f}'.format):
    print("\nConcurrent Necessity — Monthly 2024→present (vanilla thresholds)")
    print(out)

```

```python
# --- Backtest: Risk-Parity Long US Equities (SPY proxy for ES) + Long KRW (short USDKRW) ---
# Period: 2003-01-01 to present, monthly rebalanced
# Variants: (1) Always invested (Base), (2) Regime-switch (signposts on)
# Signposts: Real 10y >= +1%, Broad USD 6m momentum negative, VIX < 25
# Outputs: metrics table + single matplotlib chart (no seaborn, no custom colors)

# If needed, install:
%pip install pandas numpy yfinance pandas_datareader matplotlib scipy


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew

import yfinance as yf
from pandas_datareader import data as pdr

# ------------------------------- Config -------------------------------
START = "2003-01-01"
END   = None  # None = today

ANNUAL_TARGET_VOL   = 0.10   # 10% annualized portfolio vol target
ROLL_WIN_MONTHS     = 12     # inverse-vol lookback window (months)
MIN_OBS             = 24     # min months before we start allocating

# Regime signposts
REAL10Y_MIN_PCT     = 1.0    # DGS10 - T10YIE >= +1.0%
USD_MOM_LOOKBACK_M  = 6      # broad USD 6m momentum negative
VIX_CUTOFF          = 25.0   # avoid acute risk-off

# Risk-free series for Sharpe/Sortino (annualized %)
RF_SERIES = "TB3MS"          # FRED 3M T-bill

# ---------------------------- Helper funcs ---------------------------
def monthify(df, method="last"):
    if method == "last":
        return df.resample("M").last()
    elif method == "mean":
        return df.resample("M").mean()
    else:
        raise ValueError("method must be 'last' or 'mean'")

def log_ret(series):
    return np.log(series).diff()

def ann_factor(freq="M"):
    return 12.0 if freq.upper().startswith("M") else 252.0

def max_drawdown(cum):
    peak = cum.cummax()
    dd = cum/peak - 1.0
    t = dd.idxmin()
    s = cum.loc[:t].idxmax()
    return dd.min(), s, t

def sortino_ratio(returns, rf_annual=0.0, freq="M"):
    af = ann_factor(freq)
    excess = returns - rf_annual/af
    downside = excess[excess < 0.0]
    if len(downside) == 0:
        return np.inf
    downside_dev = np.sqrt((downside**2).mean()) * np.sqrt(af)
    mu = excess.mean() * af
    return mu / downside_dev if downside_dev > 0 else np.inf

def describe(returns, rf_monthly, label, freq="M"):
    af = ann_factor(freq)
    rf_monthly = rf_monthly.reindex(returns.index).fillna(method="ffill").fillna(0.0)
    excess = returns - rf_monthly
    cum = (1.0 + returns).cumprod()
    mdd, peak, trough = max_drawdown(cum)
    out = {
        "Ann. Return": returns.mean()*af,
        "Ann. Vol": returns.std()*np.sqrt(af),
        "Sharpe (rf)": (excess.mean()*af) / (returns.std()*np.sqrt(af)) if returns.std()>0 else np.nan,
        "Sortino": sortino_ratio(returns, rf_monthly.mean()*12*100, freq),  # rf_annual in %
        "Skew": skew(returns.dropna()),
        "Max DD": mdd,
        "Worst Month": returns.min(),
        "N Months": returns.shape[0]
    }
    return pd.DataFrame(out, index=[label]).T

# ----------------------------- Data fetch ----------------------------
yf.pdr_override()

# Yahoo Finance: SPY (proxy for equities futures TR), ^VIX, USD/KRW spot
spy = yf.download("SPY", start=START, end=END, auto_adjust=True)["Close"].to_frame("SPY")
vix = yf.download("^VIX", start=START, end=END, auto_adjust=True)["Close"].to_frame("VIX")
usdkrw = yf.download("KRW=X", start=START, end=END, auto_adjust=True)["Close"].to_frame("USDKRW")

spy_m    = monthify(spy, "last")
vix_m    = monthify(vix, "last")
usdkrw_m = monthify(usdkrw, "last")

# FRED: Broad USD (DTWEXBGS), 10y nominal (DGS10), 10y BE (T10YIE), RF (TB3MS)
twd     = pdr.get_data_fred("DTWEXBGS", START, END).rename(columns={"DTWEXBGS":"USD_Broad"})
dgs10   = pdr.get_data_fred("DGS10", START, END).rename(columns={"DGS10":"DGS10"})
t10yie  = pdr.get_data_fred("T10YIE", START, END).rename(columns={"T10YIE":"T10YIE"})
rf_3m   = pdr.get_data_fred(RF_SERIES, START, END).rename(columns={RF_SERIES:"RF_3M"})

usd_broad_m = monthify(twd, "last")
dgs10_m     = monthify(dgs10, "last")
t10yie_m    = monthify(t10yie, "last")
rf_m        = monthify(rf_3m, "last")

# Real 10y (in percentage points): nominal - breakeven
real10y = (dgs10_m["DGS10"] - t10yie_m["T10YIE"]).to_frame("REAL10Y")

# -------------------------- Component returns ------------------------
# Monthly log returns → simple returns
r_spy = log_ret(spy_m["SPY"]).dropna()
r_krw = -log_ret(usdkrw_m["USDKRW"]).dropna()  # long KRW = short USDKRW

rets = pd.concat([r_spy, r_krw], axis=1, join="inner")
rets.columns = ["SPY", "KRW"]
rets = np.exp(rets) - 1.0  # simple returns

# Align risk-free to returns index; RF_3M is annualized percent; convert to monthly simple (approx)
rf_mo = rf_m.reindex(rets.index)["RF_3M"].fillna(method="ffill").fillna(0.0) / 12.0 / 100.0

# ---------------------- Inverse-vol risk parity ----------------------
roll_vol = rets.rolling(ROLL_WIN_MONTHS).std()
inv_vol  = 1.0 / roll_vol
w_raw    = inv_vol.div(inv_vol.sum(axis=1), axis=0)

def scale_to_target(weights_row, hist, annual_target=ANNUAL_TARGET_VOL):
    hist = hist.dropna()
    if hist.shape[0] < 6:
        return 1.0
    cov_m = hist.cov()
    w = weights_row.values
    vol_m = np.sqrt(w @ cov_m.values @ w.T)
    if vol_m == 0:
        return 1.0
    vol_a = vol_m * np.sqrt(12.0)
    return annual_target / vol_a

scales = []
for i, t in enumerate(rets.index):
    hist = rets.iloc[max(0, i-ROLL_WIN_MONTHS+1):i+1]
    wrow = w_raw.iloc[i].fillna(0.0)
    scales.append(scale_to_target(wrow, hist))
scales = pd.Series(scales, index=rets.index, name="scale")

weights = (w_raw.mul(scales, axis=0)).fillna(0.0)

# Base portfolio: sum(weight * return)
port_base = (weights * rets).sum(axis=1).rename("Base")

# ------------------------- Regime switching --------------------------
usd_broad = usd_broad_m.reindex(rets.index)["USD_Broad"].fillna(method="ffill")
real10y_m = real10y.reindex(rets.index)["REAL10Y"].fillna(method="ffill")
vix_mm    = vix_m.reindex(rets.index)["VIX"].fillna(method="ffill")

usd_mom = usd_broad.pct_change(USD_MOM_LOOKBACK_M).fillna(0.0)

regime_on = (
    (real10y_m >= REAL10Y_MIN_PCT) &
    (usd_mom <= 0.0) &
    (vix_mm < VIX_CUTOFF)
)

port_regime = pd.Series(
    np.where(regime_on.values, port_base.values, rf_mo.values),
    index=port_base.index,
    name="Regime"
)

# ----------------------------- Metrics -------------------------------
def summarize(series, name):
    return describe(series.dropna(), rf_mo, name, "M")

metrics = pd.concat([
    summarize(port_base, "Base"),
    summarize(port_regime, "Regime")
], axis=1)

display(metrics.round(4))

# ------------------------------ Chart --------------------------------
plt.figure(figsize=(10,6))
cum = (1.0 + pd.concat([port_base, port_regime], axis=1)).dropna().cumprod()
cum.plot(legend=True)
plt.title("Growth of $1: Risk-Parity Long US Equities + Long KRW (Monthly, 2003–Present)")
plt.ylabel("Index Level")
plt.xlabel("Date")
plt.tight_layout()
plt.show()

# -------------------------- (Optional) Note ---------------------------
# If you want approximate futures contract counts later:
# - ES notional ~ ES_price * $50/point (Micro ES: $5/point)
# - For KRW futures, look up the CME contract USD notional per contract and divide your $-weight by that.
# This backtest uses SPY and USDKRW spot for robustness & availability; PnL is comparable to fully collateralized futures.
```
