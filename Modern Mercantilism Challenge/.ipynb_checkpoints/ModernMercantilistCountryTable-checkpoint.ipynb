```python
COUNTRIES = {
    # ── Core set ────────────────────────────────────────
    "US": {"iso":"USA", "bis2":"US", "ytick":"IRLTLT01USM156N",
           "fxtick": None, "eqtick": "^GSPC"},
    "DE": {"iso":"DEU", "bis2":"EA", "ytick":"IRLTLT01DEM156N",
           "fxtick": "EURUSD=X", "eqtick": "^GDAXI"},
    "JP": {"iso":"JPN", "bis2":"JP", "ytick":"IRLTLT01JPM156N",
           "fxtick": "JPYUSD=X", "eqtick": "^N225"},
    "BR": {"iso":"BRA", "bis2":"BR", "ytick":"IRLTLT01BRM156N",
           "fxtick": "BRLUSD=X", "eqtick": "^BVSP"},
    "KR": {"iso":"KOR", "bis2":"KR", "ytick":"IRLTLT01KRM156N",
           "fxtick": "KRWUSD=X", "eqtick": "^KS11"},
}

# ── Add new markets in one UPDATE (no self-reference) ─────────
COUNTRIES.update({
    "GB": {"iso": "GBR", "bis2": "GB",
           "ytick": "IRLTLT01GBM156N", "fxtick": "GBPUSD=X", "eqtick": "^FTSE"},
    "CA": {"iso": "CAN", "bis2": "CA",
           "ytick": "IRLTLT01CAM156N", "fxtick": "CADUSD=X", "eqtick": "^GSPTSE"},
    "MX": {"iso": "MEX", "bis2": "MX",
           "ytick": "IRLTLT01MXM156N", "fxtick": "MXNUSD=X", "eqtick": "^MXX"},
    "AR": {"iso": "ARG", "bis2": "AR",
           "ytick": None, "fxtick": "ARSUSD=X", "eqtick": "^MERV"},
    "CL": {"iso": "CHL", "bis2": "CL",
           "ytick": None, "fxtick": "CLPUSD=X", "eqtick": "ECH"},
    "PE": {"iso": "PER", "bis2": "PE",
           "ytick": None, "fxtick": "PENUSD=X", "eqtick": "EPU"},
    "CO": {"iso": "COL", "bis2": "CO",
           "ytick": None, "fxtick": "COPUSD=X", "eqtick": "GXG"},
    "IN": {"iso": "IND", "bis2": "IN",
           "ytick": "IRLTLT01INM156N", "fxtick": "INRUSD=X", "eqtick": "^NSEI"},
    "MY": {"iso": "MYS", "bis2": "MY",
           "ytick": "IRLTLT01MYM156N", "fxtick": "MYRUSD=X", "eqtick": "^KLSE"},
    "TH": {"iso": "THA", "bis2": "TH",
           "ytick": "IRLTLT01THM156N", "fxtick": "THBUSD=X", "eqtick": "THD"},
    "TR": {"iso": "TUR", "bis2": "TR",
           "ytick": None, "fxtick": "TRYUSD=X", "eqtick": "TUR"},
})
START = '2000-01-01'

# main policy or target rate series on FRED --------------------------
POLICY_FRED_SERIES = {
    "USA": "FEDFUNDS",           # Effective Fed Funds Rate
    "EA" : "ECBREF",             # ECB Main Refinancing Operations rate
    "JPN": "JPNPOLIR",           # Bank of Japan Policy-Rate Balance
    "BRA": "IRSTCB01BRM156N",    # Selic Target Rate (OECD code)
    "KOR": "KORPOLIR",           # Bank of Korea Base Rate
}

FALLBACKS = {
    "^GSPC": ["SPY", "^SPX"],   # S&P 500 ETF, stooq clone
    "^GDAXI": ["DAX"],          # DAX ETF
    "^N225":  ["EWJ"],          # Japan ETF (USD) – rough proxy
    "^BVSP":  ["EWZ"],          # Brazil ETF
    "^KS11":  ["EWY"],          # Korea ETF
}



import warnings, pandas as pd
from time import sleep                  # simplest

warnings.filterwarnings("ignore", category=pd.errors.PerformanceWarning)

import os
os.environ["FRED_API_KEY"] = "b1b93f854d911b3cdc94df6dd9ec0365"    

import pandas as pd, numpy as np, yfinance as yf
from fredapi import Fred
import world_bank_data as wb
from io import StringIO
import requests
import datetime as dt

fred = Fred()

##Helpers

# ------------- BIS policy rate: latest level -------------
# FRED back-ups in case BIS is flaky
FRED_POLICY_SERIES = {
    "USA": "FEDFUNDS",           # Fed Funds Effective
    "EA" : "ECBREF",             # ECB Main Ref. Rate
    "JPN": "JPNPOLIR",           # BoJ Policy-Rate Balance
    "BRA": "IRSTCB01BRM156N",    # Selic Target (OECD code)
    "KOR": "KORPOLIR",           # Bank of Korea Base Rate
}


def bis_policy_rate(country_code_2: str) -> float:
    """
    Latest central-bank policy rate from the BIS WS_CBPOL feed.
    • On success → numeric value (percent)
    • On any error → NaN  and a console message
    """
    url = (
        "https://stats.bis.org/api/v2/data/BIS/WS_CBPOL/"
        f"M.{country_code_2}?lastNObservations=1&format=csv"
    )
    try:
        df = pd.read_csv(url)
        val = float(df["OBS_VALUE"].iloc[0])
        print(f"   ↳ BIS policy rate OK ({country_code_2}): {val:.2f}%")
        return val
    except Exception as e:
        print(f"   ↳ BIS failed for {country_code_2} – {e}. Using NaN.")
        return float("nan")
    
def fred_series(sid):
    """
    Return a DataFrame with the single FRED series.
    If the series doesn’t exist, log it and return an empty frame.
    """
    try:
        s = fred.get_series(sid)
        if s is None or s.empty:
            print(f"   ↳ FRED series {sid} empty – using NaN.")
            return pd.DataFrame()
        return s.to_frame(name=sid).dropna()
    except Exception as e:
        print(f"   ↳ FRED series {sid} failed – {e}. Using NaN.")
        return pd.DataFrame()
    
def yahoo_px(tick: str, max_tries: int = 3) -> pd.Series:
    """
    Download Adj Close from Yahoo with 2-level fallback:
    1. primary ticker
    2. first fallback
    3. second fallback
    Raises ValueError if *all* attempts return empty.
    """
    tried = []
    for alias in [tick] + FALLBACKS.get(tick, []):
        tried.append(alias)
        df = yf.download(alias, start=START, progress=False, auto_adjust=False)
        series = df.get("Adj Close", pd.Series(dtype=float)).dropna()
        if not series.empty:
            series.name = alias
            return series
        sleep(0.5)        # tiny pause before the next attempt

    raise ValueError(f"No price history returned for {tried}. "
                     "Check network or replace ticker.")

def wb_series(ind, iso, start_year=2000, end_year=2025):
    """
    Return a single-country World Bank series as a DataFrame with a
    year-end (Dec-31) DatetimeIndex.
    """
    ser = wb.get_series(
        indicator       = ind,
        country         = iso,
        date            = f"{start_year}:{end_year}",
        id_or_value     = "id",
        simplify_index  = True,          # <- drops the 'country' level
    )                     # => plain Series indexed by YEAR (int)

    # transform the YEAR index into a proper datetime (Dec-31)
    ser.index = pd.to_datetime(ser.index.astype(str) + "-12-31")

    return ser.to_frame(name=ind).dropna()


# ── Current-account balance (% of GDP) ─────────────
def current_account_gdp(iso):
    df = wb_series("BN.CAB.XOKA.GD.ZS", iso)
    if df.empty:
        print(f"   ↳ no CA/GDP for {iso} – NaN")
        return float("nan")
    return df.iloc[-1, 0]

# ── FX reserves (USD) ───────────────────────────────
def fx_reserves_usd(iso):
    df = wb_series("FI.RES.TOTL.CD", iso)
    if df.empty:
        print(f"   ↳ no FX-reserves for {iso} – NaN")
        return float("nan")
    return df.iloc[-1, 0] / 1e9      # convert to $ bn for readability

# ── Foreign-ownership % of market cap (from CPIS+WDI) ─
def foreign_own_pct(iso3):
    foreign = foreign_equity_holding_usd(iso3)
    mktcap  = market_cap_usd(iso3)
    if pd.isna(foreign) or pd.isna(mktcap) or mktcap == 0:
        return float("nan")
    return 100 * foreign / mktcap

# ── “CheapStock” P/E percentile (2005-now) ───────────
def pe_percentile(eq_tick):
    """
    Uses yfinance TTM P/E for the index/ETF.  Not perfect but free & scriptable.
    Falls back to NaN when data or ratio are missing.
    """
    try:
        info = yf.Ticker(eq_tick).info
        pe = info.get("trailingPE")
        if pe is None or pd.isna(pe):
            return float("nan")
        # build 20-year history cache (could be persisted to disk)
        hist = yahoo_px(eq_tick)
        # simple heuristic: assume P/E history ~ price/EPS where EPS grows 5 % p.a.
        years = (hist.index[-1] - hist.index[0]).days / 365.25
        eps0  = pe and hist.iloc[-1] / pe
        eps_series = eps0 * (1.05 ** (-(hist.index[-1] - hist.index).days/365.25))
        pe_series  = hist / eps_series
        pct = (pe - pe_series.min()) / (pe_series.max() - pe_series.min())
        return pct * 100
    except Exception:
        return float("nan")


def safe(fn, *args, **kw):
    try:
        return fn(*args, **kw)
    except Exception as e:
        print(f"   ↳ {fn.__name__} failed for {args[:1]} – {e}. Using NaN.")
        return float("nan")
        
# CPI y/y %; keep as monthly to align with yields
def trailing_cpi(iso):
    # y/y CPI in %  →  decimal
    return wb_series("FP.CPI.TOTL.ZG", iso) / 100.0


def foreign_equity_holding_usd(iso3, yr=None):
    yr = yr or dt.date.today().year - 1
    url = (f"https://dataservices.imf.org/REST/SDMX_JSON.svc/CompactData/"
           f"CPIS/A.{iso3}.TXR.A.US_D?startPeriod={yr}&endPeriod={yr}")
    try:
        data = requests.get(url, timeout=20).json()
        return float(data["CompactData"]["DataSet"]["Series"]["Obs"]["@OBS_VALUE"])
    except Exception:
        return float("nan")

def market_cap_usd(iso3, yr=None):
    yr = yr or dt.date.today().year - 1
    wb_url = f"https://api.worldbank.org/v2/country/{iso3}/indicator/CM.MKT.LCAP.CD"
    params = dict(format="json", date=f"{yr}:{yr}", per_page=1)
    try:
        data = requests.get(wb_url, params=params, timeout=20).json()
        return float(data[1][0]["value"])
    except Exception:
        return float("nan")
        
def nominal_10y(cfg):
    return fred_series(cfg['ytick'])/100          # pct → decimal

def real_10y_yield(cfg):
    nom = nominal_10y(cfg).resample("ME").last()          # 'ME' = month-end
    cpi = trailing_cpi(cfg["iso"]).rename(columns={"FP.CPI.TOTL.ZG": "cpi"})
    both = nom.join(cpi, how="inner")
    real = (both.iloc[:, 0] - both["cpi"]).to_frame("real")
    return real

def real_yield_percentile(series):
    roll_min = series.rolling(120).min()
    roll_max = series.rolling(120).max()
    return ((series - roll_min) / (roll_max - roll_min)).iloc[-1]

def us10y_corr(series):
    us = fred_series('DGS10').dropna()/100
    aligned = series.dropna().resample('W-FRI').last().join(us, how='inner')
    return aligned.iloc[-52:,0].corr(aligned.iloc[-52:,1])


def gov_debt_gdp(iso):
    df = wb_series("GC.DOD.TOTL.GD.ZS", iso)
    if df.empty:
        print(f"   ↳ no debt-to-GDP data for {iso} – using NaN")
        return float("nan")
    return df.iloc[-1, 0]

def rate_hikes_bp(iso):
    hist = te.getMarketsBySymbol(f'CB{iso}')      # central-bank policy rate series idiosyncratic
    df = pd.DataFrame(hist)[['Date','Last']].set_index('Date').astype(float)
    trough = df['Last'].expanding().min()
    return (df['Last'].iloc[-1] - trough.iloc[-1])*100    # decimal→bp

FISCAL_BALANCE_WB_CODE = "GC.BAL.CASH.GD.ZS"     # cash surplus/deficit, % of GDP

FISCAL_BALANCE_WB_CODE = "GC.BAL.CASH.GD.ZS"   # cash surplus/deficit, % of GDP

# --- replace fiscal_deficit_gdp() with this version -------------------
def fiscal_deficit_gdp(iso):
    """
    Returns the latest fiscal deficit (+deficit, −surplus) as % of GDP.
    Order of attempts:
       1) World Bank JSON-Stat
       2) World Bank classic JSON
    On any failure → NaN.
    """
    try:
        df = wb_series(FISCAL_BALANCE_WB_CODE, iso)          # default jsonstat
    except Exception:
        # Retry with classic JSON (most legacy indicators support this)
        try:
            df = wb_series(
                FISCAL_BALANCE_WB_CODE,
                iso,
                format="json",            # **new**
            )
        except Exception as e2:
            print(f"   ↳ fiscal balance fetch failed for {iso} – {e2}. Using NaN.")
            return float("nan")

    if df.empty:
        print(f"   ↳ no fiscal-balance data for {iso} – using NaN")
        return float("nan")

    # World Bank: surplus = positive, deficit = negative
    return df.iloc[-1, 0]

# 3 ── fx_stats() already guards tick is None
def fx_stats(tick):
    if tick is None:
        return 0.0, 0.0, 0.0
    px = yahoo_px(tick)
    ret1y = px.pct_change(252).iloc[-1] * 100
    ret5y = px.pct_change(252*5).iloc[-1] * 100
    vol1y = np.log(px).diff().iloc[-252:].std() * np.sqrt(252) * 100
    return ret1y, ret5y, vol1y
    return ret1y, ret5y, vol1y

def eq_real_price_change(eqtick, iso):
    px  = yahoo_px(eqtick)
    cpi = wb_series("FP.CPI.TOTL.ZG", iso) / 100
    cpi_index = (1 + cpi).cumprod()

    both = (px
            .resample("ME")      # month-end ('ME'), not 'M'
            .last()
            .join(cpi_index, how="inner")
            .ffill())
    real_px = both.iloc[:, 0] / both.iloc[:, 1]
    return (real_px.iloc[-1] / real_px.shift(120).iloc[-1] - 1) * 100

def nasdaq_corr(eq_tick):
    local = yahoo_px(eq_tick).pct_change().dropna()
    ndx   = yahoo_px('^NDX').pct_change().dropna()
    aligned = local.join(ndx, how='inner')
    return aligned.iloc[-252:,0].corr(aligned.iloc[-252:,1])

rows = []
for c, cfg in COUNTRIES.items():
    print(f"\n▶ Processing {c} …")

    # 1. Bond block ----------------------------------------------------
    print("   • fetching 10-year nominal yield from FRED")
    nom = nominal_10y(cfg)          # may be empty
    
    if nom.empty:
        print("     • no yield data – filling NaN")
        nominal_10y_yield_val = float("nan")
        real_10y_yield_val    = float("nan")
        real_yld_pctile_val   = float("nan")
        us10y_corr_val        = float("nan")
    else:
        nominal_10y_yield_val = nom.iloc[-1, 0] * 100
    
        real_series           = real_10y_yield(cfg)   # <-- function still intact
        real_10y_yield_val    = real_series.iloc[-1, 0] * 100
        real_yld_pctile_val   = real_yield_percentile(real_series["real"])
        us10y_corr_val        = us10y_corr(nom)
    
    print("   • pulling government debt-to-GDP")
    debt = gov_debt_gdp(cfg["iso"])
    deficit  = fiscal_deficit_gdp(cfg["iso"])       # ← NEW line
 
    row = {
        "CurrentAcct_to_GDP_%":    safe(current_account_gdp, cfg["iso"]),
        "GovDebt_to_GDP":          safe(gov_debt_gdp, cfg["iso"]),
        "Fiscal_Deficit_to_GDP":   safe(fiscal_deficit_gdp, cfg["iso"]),
        "Policy_Rate_%":           safe(bis_policy_rate,  cfg["bis2"]),
        "FX_Reserves_$bn":         safe(fx_reserves_usd,   cfg["iso"]),
        "Nominal_10y_Yield":       nominal_10y_yield_val,
        "Real_10y_Yield":          real_10y_yield_val,
        "Real_10y_Yield_Percentile": real_yld_pctile_val,
        "Yield_US10y_Corr_1y":     us10y_corr_val,
        "ForOwn_%_MktCap":         safe(foreign_own_pct,  cfg["iso"]),
        "P_E_Pctile_2005+":        safe(pe_percentile,    cfg["eqtick"]),
        "Nasdaq_Corr_1y":          safe(nasdaq_corr,      cfg["eqtick"]),
    }

    row["Country"] = c          # <-- put this right before rows.append(row)
    rows.append(row)
    print(f"   ✓ finished {c}")

print(rows)
```

```python

from pandas.api.types import is_scalar          # ← NEW import

# turn the list of dicts into a table
df = pd.DataFrame(rows).set_index("Country")

print(df.head())  

# ------------------------------------------------------------------
# 1️⃣  Hard-cast every cell: scalar → float, else → NaN
# ------------------------------------------------------------------
def _num_or_nan(x):
    return float(x) if is_scalar(x) and pd.notna(x) else np.nan

with warnings.catch_warnings():
    warnings.simplefilter("ignore", FutureWarning)
    df_numeric = df.applymap(_num_or_nan)

# ------------------------------------------------------------------
# 2️⃣  Compute z-scores on the numeric frame only
# ------------------------------------------------------------------
z = (df_numeric - df_numeric.mean()) / df_numeric.std()

bond_cols = [
    "Nominal_10y_Yield", "Real_10y_Yield", "Real_10y_Yield_Percentile",
    "Yield_US10y_Corr_1y", "GovDebt_to_GDP", "Fiscal_Deficit_to_GDP",
    "CurrentAcct_to_GDP_%", "FX_Reserves_$bn",
    "Policy_Rate_%"
]
equity_cols = [
    "ForOwn_%_MktCap", "P_E_Pctile_2005+",
    "Nasdaq_Corr_1y"
]

desired_order = [
    "CurrentAcct_to_GDP_%", "GovDebt_to_GDP", "Fiscal_Deficit_to_GDP",
    "Policy_Rate_%", "FX_Reserves_$bn",
    "Nominal_10y_Yield", "Real_10y_Yield", "Real_10y_Yield_Percentile",
    "Yield_US10y_Corr_1y", "ForOwn_%_MktCap", "P_E_Pctile_2005+",
    "Nasdaq_Corr_1y", "Bond_Total_Score", "Equity_Total_Score"
]
df_numeric = df_numeric.reindex(columns=desired_order)

df_numeric["Bond_Total_Score"]   = z[bond_cols].mean(axis=1)
df_numeric["Equity_Total_Score"] = z[equity_cols].mean(axis=1)

# ------------------------------------------------------------------
# 3️⃣  Style only the numeric columns
# ------------------------------------------------------------------
num_cols = df_numeric.columns  # now every column is numeric or NaN

styled = (
    df_numeric.style
             .format({col: "{:.1f}" for col in num_cols}, na_rep="–")
             .background_gradient(subset=bond_cols + equity_cols,
                                  cmap="RdBu_r", vmax=2, vmin=-2)
)

print("\n✔️  All countries processed – DataFrame ready")
styled
```
