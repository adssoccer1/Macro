```python
import requests
import pandas as pd
from io import StringIO

import matplotlib.pyplot as plt
import numpy as np

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def plot_diverging_stacked_bars(df_japan):
    """
    Create a diverging stacked bar chart of Japan’s net U.S. sales by asset class,
    from 2023-04 onward.
    """
    # 1) Filter rows from 2023-04 onward
    df_filtered = df_japan[df_japan['date'] >= '2023-04'].copy()
    
    # 2) Sort in ascending chronological order (just in case)
    df_filtered = df_filtered.sort_values(by='date')
    
    # 3) Select only the columns we need and rename them
    #    (We're mapping from the original DataFrame column names to user-friendly names)
    df_plot = df_filtered.loc[:, [
        "date",           # Date
        "for_lt_treas_net",
        "for_lt_agcy_net",
        "for_lt_corp_net",
        "for_lt_eqty_net"
    ]].copy()
    
    df_plot.columns = [
        "Date",           # new col name
        "USTreasuries",
        "USAgencyBonds",
        "USCorpBonds",
        "USCorpEquity"
    ]
    
    # 4) Create the figure
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # 5) We'll define a color for each category
    category_colors = {
        "USTreasuries":  "tab:blue",
        "USAgencyBonds": "tab:orange",
        "USCorpBonds":   "tab:green",
        "USCorpEquity":  "tab:red",
    }
    
    # And the order in which we stack them
    categories = ["USTreasuries", "USAgencyBonds", "USCorpBonds", "USCorpEquity"]
    
    # 6) We'll iterate row-by-row to build the diverging stacked bars
    df_plot = df_plot.reset_index(drop=True)  # 0-based indexing
    x_positions = np.arange(len(df_plot))     # one bar per row
    bar_width = 0.6
    
    # We'll store a mapping (category -> bar handle) so we can build a single legend
    legend_handles = {}
    
    for i, row in df_plot.iterrows():
        pos_bottom = 0
        neg_bottom = 0
        
        # Go in the chosen category order
        for cat in categories:
            value = row[cat]
            color = category_colors[cat]
            
            if value >= 0:
                # Positive segment stacks "up" from pos_bottom
                ax.bar(
                    x_positions[i],
                    value,
                    bottom=pos_bottom,
                    width=bar_width,
                    color=color,
                    edgecolor="black",
                    linewidth=0.5
                )
                pos_bottom += value
            else:
                # Negative segment stacks "down" from neg_bottom
                ax.bar(
                    x_positions[i],
                    value,
                    bottom=neg_bottom,
                    width=bar_width,
                    color=color,
                    edgecolor="black",
                    linewidth=0.5
                )
                neg_bottom += value
            
            # For the legend: we only add a handle once per category
            if cat not in legend_handles:
                # bar(0,0) trick to create an invisible patch for legend
                legend_handles[cat] = ax.bar(
                    0, 0, color=color, label=cat, edgecolor="black"
                )
    
    # 7) Format the x-axis
    ax.set_xticks(x_positions)
    ax.set_xticklabels(df_plot["Date"], rotation=45)
    
    # 8) Labels, title, legend
    ax.set_xlabel("Month")
    ax.set_ylabel("Net U.S. Sales (Millions of USD)")
    ax.set_title("Japan’s Net U.S. Sales by Asset\n(Positive = Net Buying, Negative = Net Selling)")
    
    # Build a legend from the stored handles
    ax.legend(legend_handles.values(), legend_handles.keys(), loc="best")
    
    # Add a horizontal line at 0
    ax.axhline(0, color="black", linewidth=1.5, linestyle="--")
    
    plt.tight_layout()
    plt.show()


# Usage (assuming your final DataFrame of Japan data is "df_japan"):
# plot_diverging_stacked_bars(df_japan)

def plot_japan_holdings_and_sales(df_japan):
    # 1) Subset DataFrame to dates >= "2023-04"
    df_japan_filtered = df_japan[df_japan['date'] >= '2023-04'].copy()

    # 2) Sort by date so we plot in ascending chronological order
    df_japan_filtered = df_japan_filtered.sort_values(by='date')

    # 3) Convert your dates to a simple index for plotting
    dates = df_japan_filtered['date']
    x = np.arange(len(dates))

    # 4) Extract the relevant columns:
    holdings = df_japan_filtered['for_lt_total_pos']
    net_sales = df_japan_filtered['for_lt_total_net']

    # 5) Create the figure & primary axis (for Holdings)
    fig, ax_left = plt.subplots(figsize=(10, 6))

    # -- Left Axis: line plot of total holdings
    ax_left.plot(
        x,
        holdings,
        marker="o",
        linewidth=2,
        color="black",
        label="Total Holdings",
    )
    ax_left.set_xlabel("Date")
    ax_left.set_ylabel("Total Holdings (Millions of USD)", color="black")
    ax_left.tick_params(axis='y', labelcolor="black")
    ax_left.ticklabel_format(style='plain', axis='y', useOffset=False)

    # Make date labels on the x-axis
    ax_left.set_xticks(x)
    ax_left.set_xticklabels(dates, rotation=45)

    # -- Right Axis: bar plot of net sales
    ax_right = ax_left.twinx()
    ax_right.set_ylabel("Net U.S. Sales (Millions of USD)", color="black")
    ax_right.tick_params(axis='y', labelcolor="black")
    ax_right.ticklabel_format(style='plain', axis='y', useOffset=False)

    # Determine bar colors (green for positive, red for negative)
    bar_colors = ["green" if val >= 0 else "red" for val in net_sales]
    ax_right.bar(
        x,
        net_sales,
        color=bar_colors,
        alpha=0.7,
        width=0.6,
        label="Net Sales"
    )

    # 6) Title and legend
    plt.title("Japan’s Total U.S. Securities Holdings & Net U.S. Sales (2023-04 onward)")

    # Combine the legends from both axes into one
    lines_left, labels_left = ax_left.get_legend_handles_labels()
    lines_right, labels_right = ax_right.get_legend_handles_labels()
    ax_left.legend(lines_left + lines_right, labels_left + labels_right, loc="best")

    plt.tight_layout()
    plt.show()


# Example usage (assuming your final DataFrame is named df_japan):
# plot_japan_holdings_and_sales(df_japan)
def get_japan_data_from_treasury_txt():
    url = "https://ticdata.treasury.gov/resource-center/data-chart-center/tic/Documents/slt_table1.txt"
    response = requests.get(url)
    text_data = response.text
    
    # We know line 8 is the real column header row (based on the printed output)
    # so we skip the first 8 lines (lines 0..7). Then line 8 becomes header=0.
    df = pd.read_csv(
        StringIO(text_data),
        sep="\t",          # The file is tab-delimited
        skiprows=8,        # Skip lines 0 through 7
        header=0,          # Treat line 8 as the header
        na_values=["n.a."] # Convert "n.a." to actual NaN
    )
    
    # Filter rows where country == 'Japan'
    df_japan = df[df['country'] == 'Japan']
    
    return df_japan

if __name__ == "__main__":
    japan_df = get_japan_data_from_treasury_txt()
    #print(japan_df)
    plot_japan_holdings_and_sales(japan_df)
    plot_diverging_stacked_bars(japan_df)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def plot_japan_treasury_net_sales(df_japan):
    """
    Plot Japan’s net U.S. Treasury sales (for_lt_treas_net) as a bar chart
    from 2023-04 onward. Positive values are green, negative values are red.
    """
    # 1) Filter the DataFrame
    df_filtered = df_japan[df_japan['date'] >= '2023-04'].copy()
    
    # 2) Sort by date so the chart goes in ascending chronological order
    df_filtered = df_filtered.sort_values('date')
    
    # 3) Extract the columns we need
    dates = df_filtered['date']
    net_sales = df_filtered['for_lt_treas_net']
    
    # 4) Create numeric x-positions
    x_positions = np.arange(len(dates))
    
    # 5) Determine bar colors (green for >= 0, red for < 0)
    bar_colors = ["green" if val >= 0 else "red" for val in net_sales]
    
    # 6) Make the plot
    fig, ax = plt.subplots(figsize=(10, 6))
    
    ax.bar(
        x_positions,
        net_sales,
        color=bar_colors,
        edgecolor="black",
        alpha=0.7,
        width=0.6
    )
    
    # Format x-axis with dates
    ax.set_xticks(x_positions)
    ax.set_xticklabels(dates, rotation=45)
    
    # Labels and title
    ax.set_xlabel("Month")
    ax.set_ylabel("Net U.S. Treasury Sales (Millions of USD)")
    ax.set_title("Japan’s Net U.S. Treasury Sales\n(Positive = Net Buying, Negative = Net Selling)")

    # Horizontal line at 0 for reference
    ax.axhline(0, color='black', linewidth=1.5, linestyle='--')
    
    plt.tight_layout()
    plt.show()


def get_japan_data_from_treasury_txt():
    url = "https://ticdata.treasury.gov/resource-center/data-chart-center/tic/Documents/slt_table1.txt"
    response = requests.get(url)
    text_data = response.text
    
    # We know line 8 is the real column header row (based on the printed output)
    # so we skip the first 8 lines (lines 0..7). Then line 8 becomes header=0.
    df = pd.read_csv(
        StringIO(text_data),
        sep="\t",          # The file is tab-delimited
        skiprows=8,        # Skip lines 0 through 7
        header=0,          # Treat line 8 as the header
        na_values=["n.a."] # Convert "n.a." to actual NaN
    )
    
    # Filter rows where country == 'Japan'
    df_japan = df[df['country'] == 'Japan']
    
    return df_japan

if __name__ == "__main__":
    japan_df = get_japan_data_from_treasury_txt()
    plot_japan_treasury_net_sales(japan_df)
```
