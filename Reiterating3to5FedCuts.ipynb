# Reiterating 3-5 Fed Cuts over the next 12-14 Months

At the start of the year, [I recommended a long SOFR Z5 trade](https://github.com/adssoccer1/Macro/blob/main/DollarStrengthLeavesEquitiesVulnerable.ipynb), anticipating that the then-current price of roughly 96 would move closer to 97 as the market priced in approximately 125 basis points of cuts through the end of the first quarter of 2026. This call was driven by policy uncertainty surrounding the new Trump administration and an all-time high in U.S. dollar assets, with the greatest foreign exposure to these assets in history. Since then, there has also been a marginal slowdown in hard economic data, primarily due to the significant increase in the term premium on the 10-year U.S. Treasury, which began rising in September after the Fed initiated its first 50-basis-point cut of 2024.

While the anti-growth policies [I outlined in my previous pieces]((https://github.com/adssoccer1/Macro/blob/main/DollarStrengthLeavesEquitiesVulnerable.ipynb)) are now firmly in the market’s focus, this piece centers on the hard economic data that is weakening at the margins.

Trump’s broader anti-growth policies—specifically tariffs, reductions in fiscal spending, immigration restrictions, uncertainty around the Tax Cuts and Jobs Act (TCJA), and DOGE—laid the foundation for this thesis.

However, in this piece, we explore a collection of softening economic data points, including retail sales, wage growth, state and local government spending, and employment data. In aggregate, I recommend doubling down on the idea that the Fed will initiate three to five more cuts over the next 12 to 14 months. Even though SOFR Z5 now sits 30 basis points higher (96.3) than it did two months ago, the 96-97 December call spread still offers a risk 1, make 2 risk-reward (R:R) opportunity.

While many argue that the Fed is on pause due to the potential inflationary impact of Trump’s tariffs, I believe it is more likely that these tariffs will actually be dovish. Tariffs act as a friction and an anti-growth impulse to the economy, which would likely prompt the Fed to cut rates more than currently priced in this year to stimulate growth.

Deteriorating growth relative to high expectations is, of course, bearish for equities. However, shorting equities is tricky because the Fed still has 400 basis points of cuts at their disposal. Similar to the summer 2024 growth scare, an accommodative Fed could once again prevent a significant downturn in equities. When expressing a view, one should trade as closely to the underlying thesis as possible. In the case of a growth slowdown, SOFR is a more direct expression of this thesis, whereas an equities trade involves additional variables.

### Wage Growth

In 2022, many incorrectly predicted a recession. A defining characteristic of the economy that was overlooked was the upward momentum in wage growth, which sustained strong consumer spending and reflexively perpetuated further wage increases. This reflexive dynamic is no longer in place, as wage growth has been in a downward trend since 2023.

The Atlanta Fed Wage Growth Tracker suggests that wage growth continues to cool. Once it falls below 4%, it will return to its pre-COVID range. However, markets have been pricing in continued economic strength, but can no longer lean on sustained 4%+ wage growth.


[link](https://www.atlantafed.org/chcs/wage-growth-tracker)

## State and Local Spending

Much of the market’s focus remains on the evolving fiscal picture, particularly surrounding DOGE, TCJA, and discussions about the federal deficit. However, state and local government spending, which exploded at the end of 2022, is already plateauing. This is already a negative growth impulse because in order for overal growth to sustain a certain level, all else being equal, the fiscal impulse would need to remain the same. ### Employment

While U-3 unemployment remains at a healthy 4.0%, a well-known theme in today’s labor market is the increasing difficulty of finding new work. This is reflected by elevated continuing claims, low quits rates, and declining JOLTS data.

The current employment environment suggests that only a small shock is needed before layoffs accelerate, putting significant upward pressure on the unemployment rate.

We can already see initial claims and continuing claims beginning to tick up again. Perhaps this is, in part, a result of DOGE-related budget cuts in Washington, D.C.?
