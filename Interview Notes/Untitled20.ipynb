

The most important macro factors at this point in time are 
1) The fiscal factor 
2) The AI factor

There is a shift and divergence between profit growth and labor market growth. Historically poor employment is bad for markets. 

Profit margins tend to weaken first - and then we see the labor market weaken in a reinforcing negative cycle. The income driven cycle... which needs to be broken by the Fed or gov at some point and then the cycle halts. 

Whats happening right now is that profit margins overall are on there highs and yet employment has been weakening. So the weakening in the employment is not because of weaker profits. This is because we dont need as much employment in the past to produce as much profit. 

So as long as we see profits tick up - the weakening labor market is not so concerning. Employment can stay weak for a while as long as margins are going up - but it cant go negative. Its a delicate balance. 

When we have 6% fiscal deficits in an expansion - this creates inflationary pressure. The Feds best shots at stable inflation is a productivity boom - and AI is the only way right now to get a big deflationary support. 

Right now the market is saying that the productivity boom could be coming. And this is crucial because it
1) puts a ceiling on inflation
2) allows the Fed to cut more
3) Gives reason for multiples to stay elevated.
4) Crucial to allowing the gov to grow out of the debt

So markets think AI can drive a productivity boom and it certainly matters to markets whether it can or not. 

The history of producitivity booms is actually not that you get a huge boom. 

Theres a difference between captial productivity and labor productivity. We had strong margins over the last 20 years which were driven by capital productivity - but now we have labor productivity gains. 

We also had the cloud over the last 20 years which increased productivity which helped software companies boom. 

We also had the shale boom - but alot of energy companies actually went bust at this time  - and this is because the gains actually went to the consumer - not the profit margins of these companies. 

The recipient of productivity depends on the structure of the industry - if its a normal competitive industry then margins will get competed away and go to the consumer. 

The last 40 years we have seen major and consistent increases on ROIC and Margins in US stocks. This is due to 4 main reasons:
1) The first one was moving to a more capital light economy - meaning capex as a percent of profits continually fell. This process seems to be kind of over as companies are spending capital and capex a lot more on the AI build out - maybe we will return to capital light economy after thats over.
2) Corporate taxes have decreased for decades, but this is likely behind us. Its wont be politically possible due to the K shaped economy. There is still the big beautiful bill which will be the last corprate tax cut probably for a while.
3) The debt service burden is the third. Every year we saw lower and lower interest rates which decreased companies debt service burdens.
4) Falling labor costs as a percent of profits has also been falling - and now its easy to look at this chart and ask how much lower can this go? And this is where AI comes in. For the broader economy to continue to see margin expansion we really need to see labor costs as a percent of profits to continue to fall.


So if AI is going to increase labor productivity will it manifest by higher unemployment and static overll profit but higher margins - or will everyone have a job but be producing way more?

Its hard to predict this question so instead of predicting - understand what you need to watch to know what is happening and to see waht direction we are heading. The above four things are what youve needed to watch over the last 40 years - now its just that one - falling labor costs.

Fiscal is also keeping interest rates higher than it should be so if you are an industry that does not get the benefit of fiscal - all you get is the negative of higher interest rates. 

Small businesses and housing are great examples of this - these areas really got limited help from fiscal but definitely got his by high rates. 

There is kind of three neutral rates: small business, housing, and AI. The nuetral rate of small business and housing is like at least 100 bps below where we are now.
