Repo markets are seeing consistently widening spreads which indicates reserve scacity. Michael says SOFR-fed funds rate spread widening is indicating that the fed will need to do full on QE by the end of the year. 

Michael covers his usual ATI from yellen and the running down of the RRP to do hidden QE. Debt to liquidity ratio vs Debt to GDP. The termed out debt from covid is coming back to haunt the market in a debt maturity wall. 

The struggle to refinance the maturity wall will be a balance sheet problem in addition to being a interest rate problem!! Which is why the fed will need to expand its balance sheet to hold debt on their balance sheet. 

**There is static balance sheet growth which is creating tension relative to debt growth. If balance sheets are restricted (SLR ratio). Debt grows at 7-8% per year (how to get this?) so balance sheets must grow at the same rate. Balance sheets tend to grow at nominal GDP. So the spread between debt growth and balance sheet growth must be covered by the Fed's balance sheet.**

Liquidity and debt growth need to match or there are tensions. The Gov interest bill is around 4% considering the wam and that combined with the deficit which is like 7% but assume its 4% then debt is growing at 4 + 4 = 8%. So liquidity needs to match that growth rate. 

Collateral comes in two forms: public debt and shadow banks. The credit system is migrating more into the shadow bank sector as they search for different credit providers because the commercial banks balance sheets are congested.
