When you buy something in the market, you often buy from a market maker, or dealer, on the other side of your trade. Market makers make a market by offsetting their risk somehwere else to make the market. Market makers have to rebalance their deltas, vol, vega, theta, ect. The dealers is the otherside of whatever the buyside is doing. They are warehousing the risk. So if you understand how the otherside of the trade has to manage their risk that makes the markets for the buyside, then you have an edge. 

Banks also sell structured products that have to offset risk. Dealers is the phrase to capture the group of people that warehouse the majority of the risk in the market. So they take the other side of the buy side. So if you understand how the dealers are positioned, you can understand you they have to manage their risk and stay hedged. 

The options world is the risk management part of the world where people hedge. Vann charm gamma are all rehedging flows that dealers have to do. 

10-50% of flows are driven by dealers needing to rehedge. 

What percent of flows are driven by dealers managing risk? Well it depends on how much volume it being traded. It also depends on trend/mmomentum. In march of 2020 to jan of 2022, we saw a secular trend up and a massive speculation of calls with people speculating in tech and meme stocks. And also index puts on the instituational side as they were hedging as the market continued to rally. So broadly this was a very structured market in a trend for 2 years. But then what happened? Well the market came out of trend and the call speculation came off and because everyone was hedged on the index on the put side, there have been a lot of put sales. So now we are getting a different type of crowding for selling puts and selling stock. 

This is what leads to cycles. Thers a phenomenon where Vol and skew gets very bid. Customers are long puts and dealers are short them and as you decline vol gets very compressed into a decline. Then everybody liquidates their puts because they didnt work into a big decline, like they wanted them to, and eventually that sows the seeds for a big vol event. Because the next time, people dont hedge with long vol. Then this restarts the cycle, because after a vol event, people start going long vol again and the next time the decline is a step step step, way more gradual. But then the next time theyll see their puts dont work and so the cycle restarts. 

Going into covid 2020 vol was severely underbought and so thats why the opex decline after febuary was so dramatic. The covid V shape was really a vol event. Similar to august 2024. 

Looking at dealer positioning is complicated. You need to track all options flows, and open interest alone is not enough, some if it can be customer to customer flows. So there is an edge in aggregating the corect dealer flwos. 

If you know how 50% of the flows are positioned, then you can model a distribution of outcomes with certain probablilities with that flow. That is edge. 

Options markets have a reflexive affect on whether the bets that are placed in the options markets actually occur in the real world. 

If we look at brexit, we see people were hedged for a worst case outcome which was if brexit was voted for. In preparation people bought lots of out of the money puts in preparation for a worse case outcome. The event vol leading up to the vote was incredibly high relative to everything around it. So then when the event passes, regardless of the outcome, all those puts becaome very expensive leading up to the event as people buy them more and more, so there is a very high bar for what needs to happen for them to get in the money. Most importantly, if these puts dont go in the money, then dealers that are short stock, have a bunch of stock to buy back. 

**So when the event passes, regardless of the outcome, there is a massive amount of, buying of delta, selling of options, which leads to more vanna compression which leads to more buying of delta and kicks off this big loop.**

This loop happened during brexit and happened during the trump election. In 2016 we saw night of the trump election had a dramatic V dip and recovery because of these exact dynamics. Same with 2020 election we saw a contested election where we got a little dip and then off to the races. 

In Feb 2020 every body was short the march expiration but long the Feb expiration. We knew about covid in january, we had been talking about it for months, but markets had not moved. The answer is yes people knew about what was happening in China, but the markets were hedged, so it did not decline. 

**But then as soon as feb opex 2020 expired, everybod was short the march. When you get to march opex, theres an opposite effect, because there is a massive liquidation of puts and the VOL curve is inverted. 

When vol collapses, there is a massive vanna push and delta buy back, and so the whole thing drives a V bottom. 


Dealers are naturally short put and long calls because they take the othersie of the customer and pension fund flows that buy puts and sell calls to hedge. That means when we refer to vanna (decreases in IV) and charm (decreases in time) flows we are refering to the dealers, that are almost always short puts and long calls which means dealers are hedged by being short delta. So when vanna or charm occurs in means then need to BUY delta which pushes underlying prices up. 

During a window of weakness the natural buy back of delta, buy dealers, from vanna and charm forces, slows. These flows are then no longer supporting the markets, which creates windows of weakness. So windows of weakness are not really windows of weakness, its more like a window of non-strength. Its just the absence of normal vanna and charm flows from dealers that are usually there to support the markets. 

The option is not the underlying, rather the price of the underlying is the culmination of all the options open interest on the underlying.

Fundamentals of a stock hardly matter on the short term. What actually matters is how many buyers and sellers there are of the delta, and dealer positioning is a bug reflexive tools to figure out this supply and demand. In the background, longer term the fed, liquidity, fiscal spending, earnings growth, matters for the longer term. Fundemntals matter a lot less.
