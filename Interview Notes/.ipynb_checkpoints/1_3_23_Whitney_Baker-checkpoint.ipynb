On the topic of deglobalization: globalization has been taken to an extreme, especially after covid. Most of the beneficiaries of globalization have been large multinationals who can pay less on labor abroad and also benefited from strong consumption domestically. 

Now companies need to reorganize their supply chains and cost structures. Plus countries are less friendly. Corporate spending priorities are responding to consumer demand, but also responding to the shifting order of previous dependencies on China and Mexico. 

The problem with inflation in 2023 was we got in a self reinforcing cycle of higher income higher spending higher sales higher wages higher income. The only way to choke this off was with a real income squeeze or an interest rate squeeze (basically increase real rates). Unlike Europe and the UK which had a real interest rate squeeze, the US interest rate increase was not impactful. 

We need a recession and a half just to get the labor supply demand and consumption supply demand back to normal.

Post GFC, the US was the only game in town and the only place where rates got off the ground and so the dollar got a lot of inflows. Plus, combined with the QE going on the players that had their bonds bought by the FED then needed to find equivalent yeild and so pushed their former treasury investments out to something else on the risk curve. 

Now you have the Fed saying the are going to hold less of these assets, shrink their balance sheet, and the rest of the system, does not have the balance sheet capacity to hold all of these assets at current extended valuations so valuations are rerating. 

The things that did the best when the flows were flowing are now doing the worst when the flows are reversing. 

The other problem is the earnings issue. 

**Money and credit gives you purchasing power. AND purchasing power leads actual spending, because how can you spend without purhcasing power? Then the spending leads to price movement in the respective categories. So if you are looking at last months prices (inflation) then you arent really skating where the puck is going! You need to monitor what leads to price changes which is purchasing power! In 2023 we can see there is credit creation AND there is a reflexive nominal income growht cycle intact so spending will likely remain strong! and inflation sticky!!!!!**

EMs had not recieved a bid since 2013 and 2014. EM is like 20 countries and within the basket there are exporters, importers, commidity exporters/importers, goods exporters/importers. They all have different exposures to different global conditions. 

In the most recent cycle, north asia (japan korea taiwaan china) and the US were leading in a tech dominated manner and EMs had not received any flows since 13-14. 

In the GFC EMs had clean balance sheets because they had just had a crisis. So when the FED starts printing a bunch people put a lot of money into BRICS because thats what had been working. So in 2010-13 there was a huge influz of flows into risky EMs like turkey and brazil. Commodity prices were nearing all time highs and brazil running record deficits and huge over valuation of the currency. 

So in 2014 when QE switched to QT the dollar rallied and the flows to EMs stopped. 

The dollar moving creates a lot of wrong narratives. Like double check moves against inflows on fixed income and yeilds next to large dollar moves. As soon as commodity prices exploded, all commodities are essentially setteled in dollars, there was a huge demand for dollars to settle physical commodity trade. Then that faded. Then once those prices hit the commoidities importers economies like UK eruope japan and north asia all got a huge Balance of payments shock and so those are the currencies that are down, rather than the dollar going up. 

In asia there has been a lot of stress against the dollar for their currencies because they are net energy importers but they have a lot of levers they can use to manage this stress. The taiwanese lifers, the korean pensions, the japanese banks are all the big buyers of dollar assets in this cycle, particularly treasruies. So they can sell reserves to strengthen against the dollar. 

USD assets have been exposed to these positive forces for 20 years, and so now require a high bar for Baker to hold them as these forces unwind. Specifically the dollar was the currency with the highest yeild, and US equities out performance was reflively chased. Now US stocks are more than half of global market cap. And because foreigners who came running to these assets they pushed up the dollar as well. Plus a lot of these foreign buyers are unhedged in currency terms.
