Cems funds strategy started as a "skew swap". 

What is skew? IV is typically NOT the same at all strikes. Equity markets generally have a "volatility skew" or "volatility smile" where OTM puts have higher IV than OTM calls. This skew increases significatnly during market down turns. Commodity or FX markets usually have different volatility skew profiles. 

A skew swap isolates this volatility as a tradable quantity.

Skew really measures the cost of downside protection vs upside protection. The skew exists because people buy puts and sell calls to hedge. 

In 2017 there was a reflexive pinning of markets because dealers were oversupplied on index implied volatility. In 2017 we saw the lowest implied vol and realized vol in history. The index was pinned, and so when some constituents of th eindex went down or up because of their fundementals then other constituents of the index had to go up. This created dispersion.


Being able to model market makers positioning has been key for cem. Again there is obviously a significant difference between open interest which can capture customer to customer flows vs open interest that captures dealer positioning.

If everybody owns puts, the market is less likely to go down, if everyone is short puts you get a volmegeddon event like in 2018. Vanna and charm are a function of this dealer positioning. Because of the structural positioning of people that are all long the market, they are long puts and short calls because they want to hedge. This means that dealers are short puts and long calls. This means they carry a short stock position. As the volatilty collapses or as time passes and more options IV and theta decrease, then then more vanna and charm decline and the dealers can buy back deltas. 

Retail investors cant really collect the real time information to track dealer flows. Spot gamma might be able to do this? 

You need to figure out what percent of options flows are from insurance companies and pension funds that are trying to hedge by selling calls and buying puts. Like all the delta hedging is not true if the market customers are infact not structurally short calls and longs puts, making dealers short puts and long calls...
