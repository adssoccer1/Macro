People attribute skill when they have infact benefitted from luck. People want a good narrative about themselves. 

People tend to decide how much risk to take dependent on the path they have been on. So if they have been losing a lot they will take more risk to try and make up losses. If they have been wining a lot they will be more convervative

Its hard but important to de-bias yourself from these things. 

Failure to stop is a huge problem. If you dont put something like a stop loss or predetermined rules before putting on a bet then you will always find reasons to rationalize why you should keep the trade or make a move. Basically once you put the position on it is impossible not to be biased. 

Traders tell Annie that they are aware of their biases and ask themselves everyday "would i buy this position today" but they dont realize they are still being biased when they ask this question - it is almost impossible to think through and fight off you biases. 

So the only thing to do is follow pre determined rules like a stop loss.
