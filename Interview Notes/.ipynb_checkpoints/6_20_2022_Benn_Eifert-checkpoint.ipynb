Systemically risky institutions cannot trade as many risky derivatives since 2008. Most risk is now vanilla options, but there are still lots of OTC things. 

The percentage of derivatives activity is mostly in equity vanilla options. 

Retail got wiped out in 2000s tech bubble, but we saw an explosion of retail activity in covid. Short dated options became very popular because of their convexity. 

Dealers have to hedge total gamma exposure which includes the puts and calls they are long or short. If a dealer is short 10 delta calls then they have to hedge the notional exposure of those calls times the delta. 

It is extremely simplistic to say that pensions and large firms are generally selling calls and buying puts. This dynamic changes in strength throhughout different market environments. In 2022 markets were slowly grinding lower with little volatility which meant going long short dated puts was not very effective and so there was little demand for being long puts which means dealers are not as short puts as normally assumed by the cottage gamma dealer positioning crowd. 

Instead people began going long calls to the upside to try and capture a bear market rally or bounce, which changes the fundamental assumption that dealers are always long calls and short puts. 

This positioning phenomenonn could be seen in skew which shows upside is historically expensive relative to downside. 

But this all changes in different market environments. 

The VIX is very short dated measure of implied volatility. So even though the market sold off 15-20% in the first 6 months of 2022 the vix was only 20-25 and this is because at this point, realized has been low to mid 20s vol. 


A dispersion trade involves going long volatility on individual stocks within an index and simultaneously going short volatility on the index itself, betting that single-stock volatility will exceed the overall index volatility. This strategy, commonly used by institutions, exploits the diversification effect that makes indices typically less volatile than their components. Retail investors can attempt a simplified version using options on individual stocks and short-volatility positions on ETFs or indices, but challenges like high costs, complexity, and limited scalability often constrain profitability. 

Opex occurs  monthly quarterly weekly and even daily on the index. On expiration dynamic hedges have to dealer with much more aggresive delta swings (Gamma). Looking at opex purley from open interest does not mean anything. You need dealer directional open interest and then must also calculate that accross all strikes for calls and puts for all dealers. 

Gamma walls can be legititment similar to how resistance works for a technical analyst. It could be the case there is a large insitution that has a position at a certain strike, but you also dont now how this certain player is hedged already so its pretty hard to determine how they are going to dynamically hedge as price approaches the respective "gamma wall".

Even professional traders get caught selling asymettric vol that wipes them out during a black swan because markets are actually pretty efficient and it is hard to make money in quiet periods of the market. So pros will get sucked in to taking on more risk to make a little bit more money, but then get blown up when the market moves again. 

Vol markets obviously can coincide and gives hints about macro. Benn says in 2012 vol trades noticed skew in the Nikkei flattening (calls becoming more expensive relative to puts) and so vol trades started selling calls, but actually the macro was suggesting japan was about to start stimulating and so stocks were going to go up and so macro traders had been buying calls, and sure enough the Nikkei exploded higher and these vol traders got wrecked. 

There is no reason to say that a huge Vix spike is the bottom of a market. In fact you dont even need vix to hit a higher number during a large market sell off. Stocks can always grind down without spiking vix.
