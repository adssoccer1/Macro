Why is the markets down? Well we arent actually that far off the highs - the YTD returns are still high

There is 1) a rerating of the high flyers 2) there is some uncertainty about the fed because they are walking back a dec cut 

This correction does not feel game changing at this point. 

There is a new focus on affordability which has emerged and so that would be positive likely too. 

Josh is not sure if they cut in dec now or now. 

The market is still pricing 75bps of cuts next year based on two things:
1) the feds composition changing
2) the job market continuing to be weak


Josh thinks the Dec meeting is the big question mark - and not the SEP. 

Theres also now talk about the debt financing of the hyperscalers to fund the AI boom. There is more debt raising going on from the hyperscalers. The balance sheets are still strong but now they are going from exceptionally strong to normally strong. So we might see equity vol and credit spreads rising as a result. 

Josh thinks there will be an increase in layoffs and the UE rate in the next 3-6 months. He thinks there is an efficiency push going on. This will lead to lower inflation, higher unemployment, and lower rates. 

At the corporate level they are making the bet that AI will increase productivity on the tech side. So this is slowing highering on the tech side because corporates are rolling the dice of doing a little less hiring depending on AI to make up for hiring more people. 

Josh is not at all worried about inflation. He notes:
1) the labor market slowing
2) the deflationary impact of AI
3) energy prices falling
4) weak housing market

So looking at all these forward projections Josh is not at all worried about inflation but there is just some anxiety because we are above 2% inflation.
