Although there is the K shaped economy - bob sees a big diveregence between the AI mania and whats happening in the rest of the economy. 

The most evident thing here is that equal weighted s and p 500 is flat over the year, and the overall index is up a lot. And what is driving that performance is the MAG7 being driven by the AI theme, but the real economy is seemingly lagging. 

Usually as a Macro person you dont think about specific companies but because the AI story is essentially driving the aggregate market moves, the big tech earnings are becoming very important to follow. 

Front running of tariffs drove growth for the first half of the year. Then GDP growth slowed, especially when you take out the investment in software and realted ai indusrty. 

But business investment outside of AI is essentially zero. AI related spending has been the majority of Investment contribution to GDP. 

We keep seeing Hyperscaler capex expectations grow. Anlaysts keep increasing their expectations for forward looking capex. 550 billion dollars of AI capex is expected in 2026.

We are entering the max growth rate support - if the rate of growth decreases, even if it remains positive - this will be overall growth negative. 

So as this capex peaks it wont provide as much support the AI boom as much anymore. These companies are also borrowing for the first time to fund AI capex. Before they were just using freecash flow to fund AI capex, but now these cash flows are being eaten and they will start to borrow. 

So we are seeing issuance of bonds, even in the last few weeks, these companies are issuing a bunch of bonds. 

So these large companies make a ton of cash flow and they are now making an enourmous bet that the AI boom will be profitable and are essentially risking all their cash flow. 

So the question is will all of this investment pay off with a good ROI? 

Modern financial cycles end when private sector leverage gets way out over its skis and credit standards get effectively been abandoned. 

The leveraging up of the AI sector is a big signal step along that path.
