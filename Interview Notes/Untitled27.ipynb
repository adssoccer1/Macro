```python
import math
import numpy as np
import pandas as pd

def N(x):
    """Standard normal CDF"""
    return 0.5 * (1 + math.erf(x / math.sqrt(2)))

def call_bs(S, K, r, q, sigma, T):
    d1 = (math.log(S / K) + (r - q + 0.5 * sigma**2) * T) / (sigma * math.sqrt(T))
    d2 = d1 - sigma * math.sqrt(T)

    call_price = (
        S * math.exp(-q * T) * N(d1)
        - K * math.exp(-r * T) * N(d2)
    )
    return call_price

def call_bs(S, K, r, q, sigma, T):
    d1 = (math.log(S / K) + (r - q + 0.5 * sigma**2) * T) / (sigma * math.sqrt(T))
    d2 = d1 - sigma * math.sqrt(T)

    call_price = (
        S * math.exp(-q * T) * N(d1)
        - K * math.exp(-r * T) * N(d2)
    )
    return call_price

def call_approx(S, sigma, r, q, T):
    term1 = sigma * S * math.sqrt(T / (2 * math.pi)) * (1 - (r + q) * T / 2)
    term2 = (r - q) * T * S / 2
    return term1 + term2

S = 40
K = 40
sigma = 0.30
r = 0.03
q = 0.01

maturities = [1, 3, 5, 10, 20]

bs_prices = []
approx_prices = []
rel_errors = []

for T in maturities:
    bs = call_bs(S, K, r, q, sigma, T)
    approx = call_approx(S, sigma, r, q, T)
    rel_error = abs(bs - approx) / bs * 100  # percent

    bs_prices.append(bs)
    approx_prices.append(approx)
    rel_errors.append(rel_error)

results = pd.DataFrame({
    "Maturity (T)": maturities,
    "Call BS": bs_prices,
    "Call approx": approx_prices,
    "Relative Error (%)": rel_errors
})

print(results)
```
