```python
import numpy as np
import pandas as pd

# Bond setup
par = 100
c = 0.08
freq = 4
coupon = par * c / freq  # 2.0

times = np.array([0.25*i for i in range(1, 9)], dtype=float)  # 0.25,...,2.0
cash = np.array([coupon]*7 + [coupon + par], dtype=float)     # 2,...,102

def price(y):
    return np.sum(cash * np.exp(-y * times))

def duration_convexity(y):
    pv = cash * np.exp(-y * times)
    B = pv.sum()
    D = np.sum(times * pv) / B
    C = np.sum((times**2) * pv) / B
    return B, D, C

y = 0.09
B, D, C = duration_convexity(y)

dys = np.array([0.001, 0.005, 0.01, 0.02, 0.04], dtype=float)

rows = []
for dy in dys:
    B_new_D  = B * (1 - D*dy)
    B_new_DC = B * (1 - D*dy + 0.5*C*dy**2)
    B_actual = price(y + dy)

    err_D  = abs(B_new_D  - B_actual) / B_actual
    err_DC = abs(B_new_DC - B_actual) / B_actual

    rows.append([dy, B_new_D, B_new_DC, B_actual, 100*err_D, 100*err_DC])

df = pd.DataFrame(rows, columns=[
    "Δy", "B_new,D", "B_new,D,C", "B(y+Δy)", "RelErr(D) %", "RelErr(D,C) %"
])

print("B =", B)
print("D =", D)
print("C =", C)
print(df)
```
