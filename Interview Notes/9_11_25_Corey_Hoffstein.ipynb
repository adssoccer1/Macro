)[https://www.youtube.com/watch?v=znI5tYcoW6M]

Cory first tries to differentiate whether a strategy has done well because of survivorship bias or if there is an actual edge. 

Then Corey evaluates if the strategy actually makes sense - do you understand why the edge actually exists? Some TAs for example may not. 

The second thing he looks at is how you manage risk - how you bank roll this strategy. If there is a lot of risk you are taking and you might blow up then that might not be a great strategy.

Its a bad value proposition to say to investors that you will beat the market every year - because most people cannot do that and it means you will just be apologizing to clients all the time. Its better to focus you value proposition to wealth advisory or something else.
