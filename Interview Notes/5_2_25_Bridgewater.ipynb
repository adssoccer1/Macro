So what is this new environment? Modern mercantilism. The state has three goals 1) accumulate national wealth 2) accumulate geopolitical strength 3) Acheive economic self sufficiency. You can see that trade deficits reduces all three of the goals. 

So why can't trump just reverse this his policies and then we will just go back to where we were over the last 15 years? KKT gives three reasons why there is no going back:

1) The long term reason is that there are discontents with the way the system was working over last two decades, Chinas rise and globalization have gone too far. Developed countries are no longer happy with the erosion of their manufacturing policies.
2) The short term reasons are that trust takes a lot of time to build up and now this trust has been destroyed because now countries are worrying that any reliance they have on the US is now thinking about how they can reduce that dependence because they dont want to be coerced or taken adavantage of.
3) Thirdly this opens up a window of uncertainty and businesses cannot plan. The uncertainty affect cannot be put back into the tube. So no matter what trump and his team does, they cannot put the gennie back in the bottle here.

3) Countries will spend more so we will see how far the long term debt cycle can go.
4) More active government in shaping the economy fiscal policy and industrial policy.
