Goes over what tail risk hedging is and its cost of carry. 

A lot of pensions and large funds had mentally prepared for Trump being a postivie playbook for stocks. So positioning was super bullish. There was one sided positioning. But then as policy began to become clearer, it wasnt that Kris thought they were negative, but that they would be volatile and combined with crazy positioning. 

Kris personally thinks this markets will move lower and that this current rally is actually affirmation of a bear market. Some of the nastiest rallies come in bear markets. If markets slowly moved up and there was enough time for flow to exchange lower it would be different. 

Kris thinks the reflexive bid  that drove equities up is no longer here. Pension funds provided a reflexive sturcural bid that is no longer there. So now that weve bid up people will soon look to see where the pension and endowment flows are and they wont be there. 

You can look at ETF flows and lots sizes to distinguish between retail and professional instirutions. Recently we could see inflow in levered etfs and Vix etfs that are usually used by retail. Goldman also has data on this kind of retail vs institutional flows. 

The job market also has a structural bid on equities. There is such a large exposure to US stocks and people all have their 401ks bidding stocks. So as employment increases it will also hurt the structural bid for stocks, people will stop buying stocks when they lose their jobs.
