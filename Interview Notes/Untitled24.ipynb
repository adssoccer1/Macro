```python
import numpy as np

def yield_newton(price, cashflows, times, y0=0.05, tol=1e-12, max_iter=50):
    y = y0
    for _ in range(max_iter):
        disc = np.exp(-y * times)
        pv_cf = cashflows * disc
        pv = pv_cf.sum()
        f = pv - price
        if abs(f) < tol:
            break
        fp = -(times * pv_cf).sum()      # derivative dP/dy
        y = y - f / fp          # Newton step
    return y

def duration_convexity(y, cashflows, times):
    disc = np.exp(-y * times)
    pv_cf = cashflows * disc
    price = pv_cf.sum()
    dur = (times * pv_cf).sum() / price
    conv = ((times**2) * pv_cf).sum() / price
    return dur, conv

#  data
price = 101.0
cashflows = np.array([2,2,2,2,2,102.0])
times = np.array([0.5,1,1.5,2,2.5,3.0])

y = yield_newton(price, cashflows, times, y0=0.05)
dur, conv = duration_convexity(y, cashflows, times)

print("y (cc) =", round(y, 6))
print("Duration =", round(dur, 6))
print("Convexity =", round(conv, 6))
```
