A whiff of reflation. 

In the next 6 - 12 months Tian thinks we see some reflation. We are experiencing some flobal easing and liquidity creation by central banks. Savings rates are still falling and the Fed is easing into elevated GDP, so recession is unlikely.

In the US we have weak labor market data, but resilient growth and increasing liquidity. This best maps to 2002-2003 which was known as a period of "jobless recovery". 

The Fed cut in september was an insurance cut which should put a bottom under growth which should put a bottom under the Dollar and yeilds. 

Leading indicators for resilient growth in the US are tight credit spreads, low layoffs, and resilient service PMIs. 

Is the US labor market a sign of recession? Why is the job market so slow? We have hiring that is very weak, but we still have retail sales which is strong. The last time we had this happen was in 02-03. That was a jobless economoic recovery. 

Recessions are a real risk when a negative feedback loop begins with lower asset prices and then companies laying off workers and people receiving less income and then spending less. But the window of vulnerability for this to happen which was in april and may has moved on. During those times you see savings rates increase and people spend less, and we did actually see this briefly. 

But now savings rates are decreasing again and asset prices are elvated - thus recession risk is low.  

Ultimately the time to worry is when the stock market is down 5-10% and then we get a bad job print. That is when you worry about recession. 

There is moderate inflation pressure in the US. Tracking import prices is probably the most things right now. Expects elevated but not a massive resurgance in inflation. 

In China things might be starting to recover. Their 10 year has bottomed and small cap stocks have rebounded. 

Theres two main risks to focus on during periods like this. When a correction starts, all volatility begins to tick up for all assets at the same time. This is a sign of a derisking. 

Another things is the Fed cuts may be seen as political and the sign from the market is if the yeild curve has a sharp bear steapening at one of these cuts - and that will be a sign that weve kind of hit the limit for easy monetary policy and inflation fears will pick up. 

Both sides - China and the US - want supply chain independence and self reliance. So there is more central planning between governments and indusrty to reshore certain things. Both countries are trying to build a reselience from eachother. But they also dont want to have too much short term pain while doing this which is why there has been decent communication between the two countries.
