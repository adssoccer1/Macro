Joseph Wang Real Vision 3/25/25

Even without tariffs, we should expect lower growth this year. In recent past we had a huge deficit and huge inflow in immigration. Now we are shutting down immigration which will not boost growth like it did in past years. We also have a determined effort to reign in fiscal spending. There is DOGE and Howard lutnik. 

Slower popoulation growth and slower fiscal spending are both growth negative. Plus of course there are tariffs too. 

Tariffs are a politicized topic at this point. Democrats think tariffs are inflationary and republicans dont as indicated by Sentiment surveys. 

The vision of the Trump admin has been laid out to rereindustrialize the united states. We know the ammount of people employed in manufacturing and socal statistics (labour force participation) in the US has dropped. There are also national security implications like how do we fight a war if we exported all our steel manufacturing to china? 

Part of the reason all the manufacturing went to china was because of policy. China makes sure their exports are competitive though subsidies and be controlling their currency. 

The US trade deficit as ballooned to 1 trillion dollars a year in goods. Tariffs are meant to encourage consumers to buy US made goods and foreign corporations are already looking to move manufacturing into the US. 

But the companies that have benefited from globalization over the last three decades are likely to suffer some margin compression as they adjust their supply chains. Just think about apple and the iphone. 

In 2018 we saw markets fall 20% peak to trough during trumps first trade war with china. On a real economy basis in 2018 not much happened, but markets fell nonetheless, plus the Fed reacted to the falling markets and while they were signalling hikes in 2018, they actually cut multiple times. 

Wang thinks the Fed will cut more than anticipated this year. This could provide a floor to the stock market, but the market must fall before we see rate cuts. Plus lower rates is a necessary part of reshoring business in the US. Trump needs rates to be lowered. 

The goal is to move to a more stable trade balance with China. The US is the worlds largest consumer, and China is the largest manufacturer. China has been trying to produce and export its way out of their recession which requires consumer nations to buy their products. 

China also wants a more balanced economy where they rely more on domestic consumtption and not just exports. So their could be a path here where both countries are satisfied. 

In the 1980s japan was exporting a ton, especially in autos, and the solution was to get Japan to build factories in the US. Now the koreans are doing that with Hundai. We dont see chinese companies do that in US yet, but Trump said he would be open to it. The japanese made their currency too cheap in the 80s which was a complaint by the US because it made their exports cheaper. 

Wang thinks the SOFR markets may be wrong because market participants are predominantly in NYC and other liberal cities and democrats think tariffs are inflations which means no cuts.... 

Wang also notes the repatriation flows that occured after Germany annouced their fiscal packages and Europes general goal of rearming. 

Also sees China stimulus as taking flows away from US assets. 

When we consider what makes up the S&P500 we know that tech stocks make up the majority of the market, not the auto and manufacturing companies that are targeted by tariffs and the reshuffling of the global economy. 

The US dollar has strengthened against everyone because rates have been higher in the US, but as the Fed cuts later this year the dollar should weaken. But if we have a geopolitical flare up Wang expects a flight to saftey that will boost the dollar. If germany and europe rearming then this could escalate the war in ukraine. 

Vance has said that maybe being the reserve currency is not so great because it means the dollar is over valued which makes our exports too expensive which hollows out our manufacturing base, and so the Trump admin may be looking to make the dollar weaker.
