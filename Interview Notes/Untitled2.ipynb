Options dealers are exposed to adverse selection. They do not have inforamtion on the equity they are trading. Thus, they need to hedge their risk when buying the bid and selling the ask, collecting the spread, rather than taking a directional view on the position. 

Delta is pertenent in the way that it represents how many shares of the underlying the dealers need to hedge to be delta neutral. Whats more important is focusing on the greeks that affect delta.

Gamma is the sensitivity of delta to movements in the underlying price. Its useful to know the **aggregate option dealer** gamma on a stike by strike basis. The higher magnitude of gamma at each stike, the more that stike can act as an accelerant or resistance to the underlying market. A large positive gamma position would act as resistance, while a negative gamma position would act as an accelerant. This occurs because dealers are stucturally long calls and short puts which makes them long gamma. Being long gamma means they have to sell when the market goes up and buy when the market goes down. 

Vanna measure the change in delta for every one point change in annualized implied volatilty. 

Charm is the sensitivity of delta to the passage of one day of time. 

The JPM collar is the classic example of large money manager going short calls and long puts.
