So, since the Great Financial Crisis, the U.S. has sucked in a major amount of the world's capital. It was the only game in town post-GFC where interest rates got off the ground. Moreover, equity investments fueled by foreign inflows have helped equity prices increase, which have attracted more foreign inflows in a reflexive manner. And so now that global positioning is extremely invested in the United States, we run the risk of one of three potential vulnerabilities that could prick this bubble. The first being that the U.S. dollar weakens, and foreigners repatriate capital because of a weak dollar. The second being other assets outside of the U.S. start going up, and so foreigners sell some U.S. assets to reallocate to those other assets outside the U.S. And the third being that the U.S. slides into a recession, and U.S. assets fall, and interest rate differentials close, weakening the dollar's position, encouraging foreigners to sell U.S. dollar assets.

But in summary, the main concern is that if, for the last decade and a half, the U.S. has benefited from foreign inflows, reducing asset prices, encouraging further foreign inflows in a reflexive manner, that this may reverse in an unwinding, opposite reflexive manner that could become quite violent.
