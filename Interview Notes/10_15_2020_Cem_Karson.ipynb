In options you are usually transacting against a market maker(dealer). Market makers aggregate all their risk into one central book and try and hedge it all out. This is all automated in a central book. 

Delta

Vega is the dollar change in options relative to a 1% move in IV whereas vanna is the change in Delta relative to IV. 

Theta is the amount of money that comes out of an option per day that passes. 

Fix strike vol - is picking one strike price of an option chain and wathcing how IV changes at that stike over time. 

Assuming dealers are short puts and long calls and delta hedged by being short the underlying then - Everyday that gets closer to expiration the dealers have to buy delta back, thats charm. And as that volatility get compressed into expiration you have to also buy that delta back, thats vanna. To the extent that there is open interest, this dynamic happens every opex and especially every quarter on the BIG opexes. 

Vanna and charm affect 95% of market moves whereas gamma is relevant really only at the tail events or the other 5% of times. 

This about the old addage "never short a dull market" this is because of charm. Dealers are short the carry trade that is short calls and long puts and so they are long calls and short puts and then short delta, which means everyday that passes, all else constant, they have to buy back their delta hedge.... the power of charm. 

Or "markets climb a wall of worry" - this is because after a clearing event that caused high IV, the dealers have to buy back their delta due to vanna effects once that risk premium (high IV) gets pulled out of the market.
