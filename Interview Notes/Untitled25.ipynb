```python
import numpy as np

# cashflows + times
times = np.array([1/12, 7/12, 13/12, 19/12, 25/12], dtype=float)
cash  = np.array([1.75, 1.75, 1.75, 1.75, 101.75], dtype=float)

# zero curve and price
zero = 0.015 + times/(100 + np.sqrt(1 + times**2))
price = np.sum(cash * np.exp(-zero * times))

def yield_newton(price, cash, times, y0=0.1, tol=1e-12, max_iter=50):
    y = y0
    for _ in range(max_iter):
        disc = np.exp(-y * times)
        pv_cf = cash * disc
        pv = pv_cf.sum()
        f = pv - price
        if abs(f) < tol:
            break
        fp = -(times * pv_cf).sum()
        y = y - f/fp
    return y

def dur_conv(y, cash, times):
    disc = np.exp(-y * times)
    pv_cf = cash * disc
    P = pv_cf.sum()
    D = (times * pv_cf).sum() / P
    C = ((times**2) * pv_cf).sum() / P
    return D, C

y = yield_newton(price, cash, times)
D, C = dur_conv(y, cash, times)

print("Price =", price)
print("Yield (cc) =", y)
print("Duration =", D)
print("Convexity =", C)
```
