Four Fs Framework - used to describe where we are now. We arent operating in a industrial capitalism or knowledge captialism like we might have been in the past, and we arent operating in post-scarcity in an AI like economy. But the four Fs stands for 

fiscal dominance 
financial repression 
passive flows 
fiat money. 

Each concept is a stand alone concept but this is the first time that weve seen all four of these together. 

Fiscal dominance did not start after 2020, it started post GFC. The button started back then to when government spending became a significant part of GDP. Fiscal spending is now 20-25% of GDP post covid and its stuck there. We saw this in 2022 when the Fed raising rates acutally caused interst expense to go up and what turned into what was supposed to be a fiscal contraction turned into a fiscal expansion.
