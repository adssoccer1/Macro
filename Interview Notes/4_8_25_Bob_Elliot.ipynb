Bob Elliot Macro Monthly Webinar 4/8/25

Coming into 2025 in January Bob highlighted that expectations for growth ahead was becoming euphoric just at a time when we were beginning to see signs of growth moderate AND fiscal policy begin to be negative. So Bob kicked off the them "Curb your enthusiasm" to represent expectations were to high but not necessarily that growth was going to fall off a cliff. 

Over the last few weeks tariffs have out paced the negative expectations of Trumps tariffs. 

While these declines seem very large, the actual level of markets is still far away from pricing in the negative growth policies that are so far being pursued. More downside to go. 

On a relative basis the expected out performance of stocks and bonds should be zero. 

We are still in elevated levels of stocks and bonds relative to a meaningful recession. So both could continue to move significantly

While PEs have come down to around 18 this is still much higher than 15 or lower that we see when stocks bottom in recessions. Plus forward earnings growth estimates are still elevated and have not been moved down. 

Its not 4D chess, its a wrecking ball. The polcies are extremely negative and not that well thought out. There is no grand plan to balance the negative policy impulses.

There is no previous precident of tariffs like this, but in 2018 the tariffs rose prices more than the set rate because the effective tariffs flow through was higher than the set rate.

The real risk is what happens if there is a capital war. Europe especially has been critical in funding US debt. China can sell US debt and still hold the USD as USD deposits, so they wont necessarily weaken USD. 

Bob would turn bullish when he sees: a meaningful repricing of expectations, a meaningful amount of monetary and fiscal policy that is stimulative and lastly, enough momementum to create a turn. Right now we do not see any og these. Expectations are still elevated, fiscal is initiating negative policies, the Fed is sitting on its hands worried about inflation. A significant change in policy or prices could make Bob bullish.
