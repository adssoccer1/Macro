## Andy Constan Replies:
Andy Konstan wrote a response to this Carson interview, outlining why he first of all does expect a market downturn, primarily due to Trump's anti-growth policies, however he does not expect a 10-12% decline in March that will result in QE. He goes on to say why, because the Fed only does QE after lowering interest rates and the existing monetary tools like the discount window and the standing repo facility fail to stabilize markets, which means markets for one have to be destabilized by a 10-12% equity drop, which would certainly have ripple effects around the world, but does not necessarily mean that one of the credit crisis or duration crisis or liquidity crisis in the banking system would occur. It's more likely that the financial system would be able to handle a 10-12% drop in the coming months and therefore QE wouldn't even be needed. Moreover, he addresses Carson's claim that there will be a refinancing air pocket starting in Q3 2025 and continuing throughout the year as people will need to refinance the loans they made at 0% during ZURP in 2021. Konstan says this is unlikely because private balance sheets are extremely healthy. Moreover, people know that their refinancing is coming up and so they're prepared. So while this may be somewhat anti-growth and crowd out some equity investing, he doesn't see this as a crisis-inducing event.
