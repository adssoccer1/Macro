## Pretty Good Calls
I write this on Friday, April 11th, to recap markets since early March and how things played out relative to my writings in January and February.

On Globex open Sunday evening, ES put in a local bottom at 4832 and SOFR Z5 (/SR3Z25) a local high at 97.09, after Trump’s infamous tariff poster spooked markets the week prior.

On January 1st, [Monitoring Dollar Strength as a Potential Catalyst for a US Equity Selloff
](https://github.com/adssoccer1/Macro/blob/main/DollarStrengthLeavesEquitiesVulnerable.ipynb), I highlighted how U.S. dollar assets had the highest exposure to foreign ownership at any point in history, culminating in what appeared to be a blow-off top in dollar demand and equity strength at the beginning of the year. I emphasized a long Z5 SOFR 96–97 call spread, anticipating the price would move from 96 to 97 as equities sold off at some point during the year. I also highlighted how an equity sell-off could be exacerbated by the compounding effects of dollar weakness and equity weakness, giving foreigners with unhedged FX exposure heavy losses.

Not only did I sell all my equity exposure, I went long ES put spreads going into late February and early March. I closed these out for modest gains in mid-March while keeping my long SOFR call spreads. My short ES gains could have been significantly larger if I had made a couple of different decisions. On Liberation Day, it became clear the sell-off would continue in dramatic fashion, and on Sunday’s April 6th futures open, the gap-up in SOFR was dramatic enough that profits had to be taken, resulting in a 200% gain on my original long SOFR position.

At this point, I was flat and considering buying back long-term equity exposure.

One crucial lesson I already knew but relearned was that when the market is at peak fear—as it was on Tuesday—and that fear is controlled by one individual beginning to inflict self-imposed damage, it’s worth paying a cost of carry to get exposure to a potential policy reversal. Starting Monday, I opened long 0DTE calls way out of the money, anticipating a policy announcement from the President to arrest the market sell-off. There was no reversal, and my calls went to zero.

On Wednesday, I opened 0DTEs again, but for whatever reason, I decided to sell the 0DTEs I bought for 5.35 at 8 because… *shrug*—just 30 minutes before Trump’s tariff delay announcement—missing a 40x in 2 hours. I will never forget this trade… lol.




**You must know what weapons to choose when you go into battle, and you must think like the enemy!** Trump was in control of his self-imposed pain, so a small cost of carry for exposure to an asymmetric reversal at some point in the week was a no-brainer trade

## Where We Go From Here

### The Treasury Market Sell Off
The equity sell-off from late February to April 2nd was orderly. DXY was strong, as were credit spreads and the Treasury market. However, after Liberation Day revealed the severity of Trump’s intentions, rumors of capital wars and foreign boycotting/selling of U.S. dollar assets began circulating. The long end of the U.S. Treasury yield curve started moving up, and the dollar moved down.



This is precisely what we anticipated at the beginning of the year: [see here](https://x.com/aidansinclair_/status/1897728595474497783) and [see here](https://x.com/aidansinclair_/status/1897722931469226221)!

This is a classic dynamic seen in a balance of payments crisis. Which I also wrote about [here](https://github.com/adssoccer1/Macro/blob/main/BalanceOfPaymentsThoughts.ipynb).

Trump’s explicit goal is to shrink the current account deficit, thereby shrinking the capital account surplus. If there is less structural demand globally for Treasuries, then speculators will front-run the decline in demand. While I doubt China has been selling its Treasuries in large numbers since April 2nd, speculators will front-run the structural demand shift that Trump is unknowingly pushing for. It’s crucial to note that foreign central banks like China hold less than 10% of their U.S. Treasury exposure in 10-year or 30-year bonds—those are the tenors that have seen the largest downward moves this week.



Plus, a new narrative is emerging where people have yet to recognize that if treasury yields continue to rise because of foreign selling and speculators front-running the change in structural demand for U.S. dollar assets, the Fed actually cannot cut rates because the clearing price of attracting foreign capital into the treasury market is increasing, even while equities and the real economy suffer.

In my view, the sell-off in Treasuries is unlikely to persist. I give more credit to the "[basis trade](https://www.bloomberg.com/news/newsletters/2025-04-09/everything-you-need-to-know-about-the-basis-trade-spooking-markets?srnd=phx-oddlots)" and speculators than outright foreign central bank Treasury sales. Although many expect a flight to safety to bid bonds during equity sell-offs, it’s actually quite common for bonds to sell off in corrections as aggressive as this one (see 2008 and 2020).

After the bear steepener, Trump’s tariff delay announcement—likely at the insistence of Bessent—put another floor under the stock market, but has so far failed to completely arrest the long end of the Treasury yield curve.

Jamie Dimon also called for SLR relief this week to alleviate pressure on the bond market. We wrote about SLR relief [here](https://github.com/adssoccer1/Macro/blob/main/MonitoringBankBalanceSheets.ipynb)!

### The Next Trade

In terms of deciding where markets go from here, I see the same policy mix that I [discussed](https://github.com/adssoccer1/Macro/blob/main/DollarStrengthLeavesEquitiesVulnerable.ipynb) at the beginning of the year—but now with more clarity.

Decreased immigration, deficit reduction, and tariffs were, as anticipated, more significant than markets had priced in at the start of the year. These policies remain important forces driving negative growth pressures, but now we have a better sense of where Trump is willing to support markets and compromise on policy.

We cannot forget the linear decline the real economy was in before Liberation Day—[I wrote about that here](https://github.com/adssoccer1/Macro/blob/main/Reiterating3to5FedCuts.ipynb).

I would highlight that we are seeing slower non-residential business fixed investment, weaker consumer spending, lower real wage growth, and reduced state and local spending. Plus, in our hyper-financialized economy, we will see the effects of this 20% drawdown in equities (401(k)s) and the spike in the long end further impact real economic data in the months to come.

What are the positive growth impulses that will support the still elevated forward earnings for U.S. equities? Perhaps in 2H25, we will be in an environment where Trump has negotiated tariffs down and pivoted to his growth agenda  [see here](https://x.com/CGasparino/status/1910747967595634751), banking deregulation has added a bid to Treasuries, and the Fed is beginning another rate-cutting cycle. For these reasons, I tentatively believe we’ve bottomed in SPX for 1H25. But is this enough to support impressive U.S. equity market returns this year?

If you bought the ES 4832 dip Monday night, I think yes.

### How Markets Return to Normalcy

One difference post–Liberation Day is that we now face a larger left tail risk, as we must slowly balance the restructuring of the previously consistent foreign bid for U.S. Treasuries. If this bid disappears too quickly and neither a policy reversal nor the Fed can arrest a larger Treasury sell-off, U.S. markets could enter crisis territory rapidly. However, if one believes the long end will stabilize, we return to the linear movements of the real economy—movements protected by 425 basis points of potential rate cuts.

We’ve also seen 1-year, 1-year inflation expectations explode upward on the back of tariffs, while 5-year, 5-year expectations have decreased significantly. This gives the Fed leeway to cut, given their reliance on forward-looking inflation metrics. I remain of the view that tariffs are a non-inflationary tax because they represent a one-time increase in the price level, rather than an increase in the base money supply or the velocity of money. In fact, they are a short-term friction to growth. In the case of 2018, they were primarily absorbed by company profit margins, which meant fewer hires and, consequently, lower consumer income. In the longer term, reversing globalization would be inflationary—but not before the Fed cuts to support growth later this year.

So long as the sell-off in long bonds is arrested, the Fed will commence with cuts in the next 12 months. I recommend selling short-dated put credit spreads on the 10-year and 30-year bonds, as I anticipate Trump and Bessent will not tolerate a long-end sell-off that breaches a 5-handle. The Fed put also lies at a slightly lower strike.

Bessent understands that as the capital account in the United States begins to shrink, he must supplement the disappearing Treasury bid with domestic banking deregulation via the SLR and through deficit reduction. I believe he will manage the transition of the structural bid for Treasuries in an orderly fashion, allowing the real economy to resume its role in driving both equities and interest rates.
