# Exposing Dealer Gamma Exposure

Dealer Gamma Exposure, OPEX, Liquidity Cascades, Passive Investing, Leveraged ETF Rebalancing.
These are popular themes among retail and even professional investors. But do they actually offer edge? Are they a necessary component of an investor’s knowledge base? What real insights do they provide?

This piece explores the shortcomings and select strengths of Dealer Gamma Exposure. The general idea is to capture a snapshot of the  [market's order book](https://squeezemetrics.com/monitor/download/pdf/The_Implied_Order_Book.pdf). Consider the following:

- All that determines price is the supply and demand for delta.
- Dealers often make up more than 50% of the activity on an options chain.
- Dealers dynamically delta hedge their options positions by going long/short the underlying.
- Dealers provide most of the supply and demand for delta as they dynamically hedge.


There are data sources that stream a live list of all options traded on an exchange, which can be used to build models of dealer positioning. If one aggregates all these options flows across all exchanges and can determine if a dealer was a participant—and whether they were opening or closing the trade—then one can build a model that tracks dealer-directional open interest at each strike and expiry on an options chain. Once you have a model that approximates dealer-directional open interest accurately, you can calculate the aggregated delta exposure of dealers and derive their gamma exposure.

Proponents of Dealer Gamma Exposure make assumptions about how you can predict directional moves in markets based on dealer positioning. A major assumption is that dealers automatically delta hedge any option position they open. But how can one determine whether they’re hedging through the underlying—or using options themselves? That throws a huge wrench into many peoples' basic understanding of dealer gamma exposure.

Moreover, many assume the market is structurally short calls and long puts, which makes dealers long calls and short puts, forcing them to dynamically delta hedge by selling the underlying. But this isn’t always the case. In fact, markets often enter long call / long put environments. Nonetheless, many build on this faulty structural assumption and go on to predict how Vanna and Charm will affect Gamma, and how that will force dealers to buy or sell the underlying as they hedge.

*(Vanna is how delta changes with moves in implied volatility and the underlying. Charm measures how delta decays over time—specifically, with each passing day. Gamma is the second derivative of delta—how delta changes with a one-point move in the underlying.)*

For example, the November 2024 election was a binary event where the market viewed a Harris win as growth-negative and a Trump win as growth-positive. There were also concerns over market stability in the event of a contested outcome. Heading into the election, many investors were hedged via long puts—meaning dealers were short puts and hedged that exposure by being short the underlying. As the event approached, implied volatility was elevated, forcing dealers to sell more delta to hedge their short puts. But once the event passed and Trump won decisively, IV collapsed, pushing the market higher and forcing dealers to buy back delta. That’s Vanna. On top of that, each day that passed reduced the delta on dealers’ long options—shrinking their hedge requirement and forcing more delta buybacks. That’s Charm.

But wait, it sounds like the Dealer Gamma Exposure extraplatoooors were right about the election!? And they were! 

There are several shortcomings that make anticipating Dealer Gamma Exposure an art more than a science:

1)  Dealers may use options themselves to delta hedge, rather than the underlying.
2)	Correctly identifying whether a trade includes a dealer participant is difficult.
3)	Correctly characterizing the side the dealer is on—are they opening or closing, long or short—is even harder.
4)	Dealers likely have DGE models themselves to front-run dealer flows.
5)	Dealer positioning is now well-mapped market-wide.
6)	Open interest alone is nearly useless without additional context.
7)	Dealers rehedge their deltas at different times—often the following morning.
    

Together, these shortcomings create large margins of error for Dealer Gamma Exposure models that lack the nuance to overcome them. Unless dealer positioning is blatantly obvious—like during the 2024 election—many investors over-index on naive models and end up disappointed by their predictive power.

Another important dealer-related dynamic is when there’s a wide wing of concentrated directional open interest far out-of-the-money on speculative underlyings. Tesla is a prime example. Its violent move from 250 to 480 wasn’t driven by fundamentals—it was a reflexive gamma squeeze. Retail speculators were long far OTM calls, and once the move began, the feedback loop took over. It might not be so amateur to hold long OTM calls when there’s clear retail interest piling in.

**The greatest insights the Dealer Gamma Exposure idea offers is that you can find edge in creative ways modeling market structures and flows. Every beginning quant should learn the Dealer Gamma Exposure model to inspire ideas about how to source information to craft an edge.**

When one starts thinking about Dealer Gamma Exposure, it’s natural to wander into the world of vol. Skew refers to the fact that implied volatility for OTM puts tends to be higher than for OTM calls. This happens because puts are in greater demand—especially as institutional hedging flows from pensions and insurers add buying pressure to puts and selling pressure to calls. But assuming the market is always structurally short calls and long puts is a mistake.

On the Nikkei in 2012, the volatility smile—or skew—began to flatten. Many vol traders started selling calls as the IV on calls rose and made them appear expensive and an attractive opportunity for a vol seller. But this was actually an early sign that macro conditions in Japan were shifting. The country was preparing to stimulate aggressively, and macro investors began buying calls to front-run the policy response. Sure enough, the Nikkei exploded higher—and the vol traders who had sold those calls lost money.

Fixed strike vol is also extremely important. Imagine the market is drawing down and IV is elevated, yet you still believe more downside is coming and want to buy puts. If you buy when IV is already high, the market can continue falling, but in a more orderly, lower-vol manner, which causes IV to compress and leaves your puts flat or even down! When fixed strike vol is elevated, using spreads can help mitigate unwanted impacts from vol decay and collapsing skew.

Briefly diving into the world of vol will throw you into Volmageddon (2018) and March 2020. At a high level, vol can become structurally oversupplied or undersupplied. Heading into 2018, vol was structurally undersupplied—many investors had been shorting vol for years. The market was fragile, with no cushion from long vol. In March 2020, COVID hit global headlines in February, but the market kept rallying until February OPEX. After that, we saw a sharp unwind—this time due to vol oversupply, as panic hedging overwhelmed market makers and liquidity disappeared.
