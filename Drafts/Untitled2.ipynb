The general idea of Dealer Gamma Exposure is to capture the market's order book. All that determines price is the supply and the demand for delta. A key insight into market structure, is that dealers often make up more than 50% of the activity on an options chain. 

There are data sources that stream a live list of all options traded on an exchange. If one aggregates all these options flows on all the exchanges AND has a model to determine if an options trade had a dealer participant involved AND can determine if the dealer was opening or closing this trade, then one can build a model that tracks dealer directional open interest at each strike and expiry on an options chain. Once you have a model that is approximating dealer-directional open interest accurately, you can then calculate the aggregated delta exposure of dealers.

 At this point, proponents of Dealer Gamma Exposure make assumptions about how you can predict directional moves in markets based off of this dealer positioning. The first assumption is that dealers automatically delta hedge any options position they open. Overwhelmingly, I'm sure dealers do dynamically delta hedge, But how can one determine if they're hedging through the underlying or if they're hedging using options themselves! This throws a huge wrench into many's basic understanding of dealer gamma exposure.

Moreover, many assume the market is structurally short calls and long puts, which makes dealers long calls and short puts, which forces them to dynamically delta hedge by selling the underlying delta. But this is not always the case. Nonetheless, they continue to make a second-order assumption, which is how vanna and charm affect gamma and the dynamic delta hedging of the dealers.

(Vanna is how delta changes with changes in implied volatility. Charm is the measure of how delta changes with the passing of one day. And gamma is the second derivative of delta, representing how delta is changing with a one point move on the underlying. )

The November 2024 election was an event that had a binary outcome where the market saw a Harris win as growth negative and a Trump win as growth positive. Plus, there was concern over market stability in the face of a contested election. Going into the election, many people were hedged via long puts. So, dealers were short puts and hedged by being short the underlying. Moving into the event, implied volatility was high forcing dealers to sell more delta than normal to cover their short puts. Once the event passed and Trump won decisively, IV compressed, pushing up the market, forcing dealers to buy back deltas. That is vanna. Plus, everyday that passed reduced delta on the options dealers were long, likewise reducing how much they need to hedge, forcing them to buy back delta, propelling markets higher in a reflexive manner.
