# Super Cheap Lunch for September Rate Cut Decision
### How to Structure a Near Risk Neutral Bet for a 7.7% Betting on a 25 bps Cut in September.

The Federal Reserve will meet on September 18, 2025, to make its next interest rate decision. The CME FedWatch tool currently prices an 85% chance of a 25bps cut, [see here](https://www.cmegroup.com/markets/interest-rates/cme-fedwatch-tool.html). 

Using a combination of the October 2025 30-Day Fed Funds Futures contract (ZQV5) and a couple of Kalshi’s digital options contracts, it’s possible to build a trade that: ends flat if the Fed makes a 0 or 50 bps change, and makes 7.7% on invested capital if the Fed cuts 25 bps. 

Consider this snapshot of current prices on 8/15/25:

- ZQV5 = 95.90
- Kalshi's Fed Cuts 25 bps contract: Yes- 69 cents
- Kalshi's Fed cuts > 25bps contract: Yes- 5 cents



[See Kalshi's Sept Fed Funds market here](https://kalshi.com/markets/kxfeddecision/fed-meeting)

Then consider taking a short position in ZQV5 at a price of 95.90. The terminal price of this October contract will be determined by the average Fed Funds rate over the month, and there is another interest rate decision on October 29th allowing 2 days after that meeting to also impact the terminal prices of this contract, outside of just the September meeting. However this is an extremely minimal impact, so essentially the October contract will reflect the September interest rate decision. 

If there are no cuts in September, ZQV5 should settle around 95.67. If there is one cut, around 95.93, if 50 bps then around 96.18. Multiply each bps move by `$41.67` for each contract owned. 

If you pair your short 1 ZQV5 position with the purchase of 
- 1,234 of Kalshi's Fed Cuts 25 bps contract: Yes at 69 cents each 
- 2,126 of - Kalshi's Fed cuts > 25bps contract: Yes at 5 cents

Then consider the following PnL scenarios for each Fed decision:

| Fed Decision | ZQ settles at| Futures P&L from Short @95.90  | Kalshi Payout | Kalshi Cost | Net PnL     |
|---------------|-----------|--------------|---------------|-------------|-------------|
| Hold (0 bps)   | 95.67     | +\$958.41     | \$0.00         | −\$957.76    | +\$0.65      |
| Cut 25 bps     | 95.93     | −\$125.01     | \$1,234.00     | −\$957.76    | +\$151.23    |
| Cut 50 bps     | 96.18     | −\$1,166.76   | \$2,126.00     | −\$957.76    | +\$1.48      |


Say the margin required to hold one ZQV5 contract is about `$925` and the cost of the paired Kalshi contracts is `$957.76`, combining to `$1,882` needed for a `$151` return or 7.7% on total required capital.

This is a pretty unique risk reward for something the market is pricing at 85% chance of happening. 

Recognize this trade does have a large loss scenario in the unlikely case of emergency rate cuts to zero in the next 5 weeks. Add a stop to the short ZQV5 position at 96.18 so the trade stays within the 50 bps bound. I personally have a stop at 96.12. And if you are worried of such a drastic emergency that your ZQV5 stops are skipped, add some far OTM long SOFR options to your portfolio! They'll help you worry less.
